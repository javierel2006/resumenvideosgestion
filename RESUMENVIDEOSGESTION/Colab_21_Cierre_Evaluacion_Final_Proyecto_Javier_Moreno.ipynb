## Análisis y resumen del video 21

**Asignatura:** Ingeniería de Proyectos

### Propósito
Comprender cómo evaluar los resultados obtenidos, verificar el cumplimiento de los objetivos y realizar correctamente el cierre de un proyecto.

## 1. Resumen del tema

La evaluación final permite revisar si el proyecto logró lo que se había planteado.

Se pueden analizar:

- Objetivos.
- Alcance.
- Cronograma.
- Costos.
- Calidad.
- Recursos.
- Riesgos.
- Satisfacción de los interesados.
- Entregables.
- Lecciones aprendidas.

La evaluación debe comparar los resultados reales con las referencias establecidas durante la planificación.

## 2. Cumplimiento de objetivos

El primer paso es revisar los objetivos.

Ejemplo:

**Objetivo:** desarrollar una aplicación web con cinco módulos.

**Resultado:** se desarrollaron los cinco módulos y fueron aceptados.

En este caso existe evidencia de cumplimiento del objetivo.

La evaluación debe apoyarse en resultados verificables y no solamente en opiniones.

## 3. Evaluación del alcance

Se debe comprobar:

- Qué estaba incluido.
- Qué fue entregado.
- Qué quedó pendiente.
- Qué cambios fueron aprobados.
- Qué elementos fueron excluidos.

Esto permite determinar si el resultado final corresponde al alcance aprobado.

## 4. Evaluación del cronograma

Se compara:

**Fecha planificada**

con

**Fecha real**

Ejemplo:

Plan de entrega: 30 de noviembre.

Entrega real: 5 de diciembre.

La diferencia es de **5 días**.

Después de identificar la desviación se debe analizar su causa y registrar la información como parte del cierre.

## 5. Evaluación de costos

También se compara:

**Presupuesto planificado**

con

**Costo real**

Ejemplo:

Presupuesto: **$20.000.000**

Costo final: **$21.500.000**

Diferencia:

**$1.500.000**

La evaluación debe determinar por qué ocurrió la diferencia y si fue causada por cambios, riesgos, errores de estimación u otras situaciones.

## 6. Evaluación de calidad

Al finalizar se revisa si el producto cumple los requisitos y criterios de aceptación.

En software se pueden revisar:

- Funcionalidades.
- Pruebas.
- Errores pendientes.
- Seguridad.
- Rendimiento.
- Documentación.
- Aceptación del usuario.

La calidad final debe compararse con los criterios definidos para el proyecto.

## 7. Evaluación de resultados

Una evaluación sencilla puede utilizar una tabla:

| Área | Planificado | Real | Diferencia |
|---|---:|---:|---:|
| Alcance | 10 módulos | 10 | 0 |
| Cronograma | 100 días | 105 | +5 días |
| Presupuesto | $20 M | $21,5 M | +$1,5 M |
| Calidad | 95% pruebas exitosas | 97% | +2 puntos |

La tabla facilita observar rápidamente las diferencias entre el plan y el resultado.

```python
# Ejemplo de evaluación final

plan = {
    "alcance": 10,
    "dias": 100,
    "presupuesto": 20_000_000
}

real = {
    "alcance": 10,
    "dias": 105,
    "presupuesto": 21_500_000
}

print("EVALUACIÓN FINAL")
print("-" * 30)
print(f"Alcance: {real['alcance']} de {plan['alcance']} módulos")
print(f"Variación de tiempo: {real['dias'] - plan['dias']} días")
print(f"Variación de costo: ${real['presupuesto'] - plan['presupuesto']:,.0f}")
```

## 8. Aceptación final

La aceptación confirma que los entregables cumplen las condiciones acordadas.

Puede incluir:

- Revisión del producto.
- Validación de requisitos.
- Pruebas.
- Aprobación del cliente.
- Acta o registro de aceptación.

La aceptación formal ayuda a establecer que el producto fue recibido conforme a los criterios definidos.

## 9. Cierre administrativo

El cierre puede incluir:

1. Confirmar entregables.
2. Cerrar contratos.
3. Finalizar pagos pendientes.
4. Organizar documentos.
5. Liberar recursos.
6. Registrar lecciones aprendidas.
7. Elaborar el informe final.
8. Formalizar el cierre.

Estas actividades permiten finalizar el proyecto de manera ordenada.

## 10. Evaluación del desempeño

La evaluación también puede utilizar indicadores estudiados anteriormente.

Por ejemplo:

**SPI = EV / PV**

**CPI = EV / AC**

Estos indicadores pueden ayudar a complementar la evaluación del desempeño de tiempo y costo.

Sin embargo, la evaluación final debe considerar también el alcance, calidad, satisfacción y resultados obtenidos.

## 11. Satisfacción de los interesados

No basta con terminar las actividades.

También es importante conocer si el resultado respondió a las necesidades de los interesados.

Se puede obtener información mediante:

- Encuestas.
- Reuniones.
- Entrevistas.
- Retroalimentación.
- Evaluaciones de usuarios.

Esta información puede convertirse en una nueva fuente de lecciones aprendidas.

## 12. Relación con todo el proyecto

La evaluación final reúne los principales temas estudiados:

**Requisitos**
→ qué debía cumplirse.

**EDT**
→ qué trabajo se organizó.

**Cronograma**
→ cuándo debía realizarse.

**Costos**
→ cuánto debía costar.

**Recursos**
→ con qué personas y medios se realizó.

**Riesgos**
→ qué situaciones afectaron el proyecto.

**Calidad**
→ qué tan bien se obtuvo el resultado.

**Comunicaciones**
→ cómo se informó.

**Adquisiciones**
→ qué se obtuvo de proveedores.

**Valor Ganado**
→ cómo fue el desempeño.

**Agile**
→ cómo se adaptó el trabajo.

**Cierre**
→ cómo se formalizó la finalización.

## 13. Ejemplo final aplicado a software

Una empresa desarrolla una aplicación web.

- Se completaron los módulos acordados.
- El proyecto terminó cinco días después de lo previsto.
- El costo superó el presupuesto inicial.

- Las lecciones aprendidas fueron registradas.

La evaluación final debe mostrar tanto los logros como las desviaciones y sus causas.

**¿Qué debemos mejorar?**

## 16. Importancia para Ingeniería de Proyectos

La evaluación final permite conocer no solamente si el proyecto terminó, sino cómo terminó.

Un proyecto puede haber cumplido el alcance pero presentar desviaciones de tiempo o costo. También puede haber cumplido fechas y presupuesto pero presentar problemas de calidad.

Por eso la evaluación debe observar el proyecto de manera integral.

## 17. Conclusión general

El cierre representa la última etapa formal del proyecto.

Una buena evaluación final permite:

- Comprobar objetivos.
- Revisar entregables.
- Comparar plan y realidad.
- Identificar desviaciones.
- Analizar resultados.
- Registrar lecciones aprendidas.
- Formalizar la aceptación.
- Cerrar contratos y recursos.
- Preparar conocimiento para futuros proyectos.

La secuencia final puede resumirse:

**Verificar → Evaluar → Documentar → Aceptar → Cerrar → Aprender**
