## 1. Resumen del tema

La integración consiste en coordinar las diferentes partes del proyecto para que funcionen como un conjunto.

Durante un proyecto existen:

- Alcance.
- Cronograma.
- Costos.
- Recursos.
- Riesgos.
- Calidad.
- Comunicaciones.
- Interesados.
- Adquisiciones.

Todas estas áreas están relacionadas. Una decisión en una puede afectar a las demás.

La integración ayuda a mantener una visión general del proyecto.

## 2. ¿Por qué es importante integrar?

Un proyecto no funciona como actividades aisladas.

Por ejemplo:

Si cambia un requisito, puede cambiar:

**Alcance → Cronograma → Costos → Recursos → Riesgos → Calidad**

Por eso una modificación debe analizarse de manera integral antes de aplicarla.

## 3. El gerente del proyecto y la integración

El responsable del proyecto debe coordinar las diferentes actividades y decisiones.

Entre sus funciones están:

- Mantener una visión general.
- Coordinar al equipo.
- Revisar avances.
- Gestionar cambios.
- Resolver conflictos.
- Coordinar interesados.
- Revisar riesgos.
- Asegurar que las actividades estén alineadas con los objetivos.

## 4. Gestión de cambios

Los cambios pueden aparecer por:

- Nuevas necesidades del cliente.
- Problemas técnicos.
- Cambios legales.
- Nuevos riesgos.
- Cambios de presupuesto.
- Cambios de prioridades.

Un cambio no debería aplicarse de manera improvisada.

Un proceso sencillo es:

**Solicitud → Análisis de impacto → Decisión → Actualización → Comunicación → Ejecución**

## 5. Control integrado de cambios

Antes de aprobar un cambio se pueden revisar:

- Alcance.
- Tiempo.
- Costo.
- Calidad.
- Recursos.
- Riesgos.
- Contratos.
- Interesados.

Ejemplo:

El cliente solicita una nueva funcionalidad.

Antes de aceptarla se debe determinar cuánto tiempo adicional requiere, cuánto cuesta y qué otras actividades pueden verse afectadas.

## 6. Documentación del proyecto

La documentación permite conservar información importante.

Puede incluir:

- Requisitos.
- Planes.
- Cronogramas.
- Presupuestos.
- Cambios aprobados.
- Informes.
- Resultados de pruebas.
- Contratos.
- Entregables.
- Lecciones aprendidas.

La documentación facilita demostrar qué se realizó y cómo se tomaron las decisiones.

## 7. ¿Qué significa cerrar un proyecto?

Cerrar un proyecto significa completar formalmente las actividades y confirmar que el trabajo ha terminado.

El cierre puede incluir:

1. Verificar los entregables.
2. Obtener aceptación.
3. Cerrar contratos.
4. Organizar documentos.
5. Liberar recursos.
6. Registrar lecciones aprendidas.
7. Elaborar el informe final.
8. Formalizar el cierre.

## 8. Aceptación de los entregables

Antes del cierre es necesario comprobar que el producto cumple lo acordado.

Por ejemplo, en un proyecto de software:

- Las funcionalidades deben estar terminadas.
- Las pruebas deben haber sido realizadas.
- Los requisitos aceptados deben estar cubiertos.
- La documentación debe estar disponible.
- El cliente debe validar la entrega.

Esto permite evitar cerrar un proyecto con actividades importantes pendientes.

## 9. Lecciones aprendidas

Las lecciones aprendidas registran lo que funcionó y lo que podría mejorarse.

Algunas preguntas útiles son:

- ¿Qué salió bien?
- ¿Qué salió mal?
- ¿Qué problema se repitió?
- ¿Qué decisión ayudó al proyecto?
- ¿Qué debería hacerse diferente?
- ¿Qué recomendación puede servir para otro proyecto?

Estas experiencias pueden convertirse en conocimiento para futuros proyectos.

```python
# Ejemplo sencillo de resumen final del proyecto

proyecto = {
    "entregables_completados": 10,
    "entregables_pendientes": 0,
    "riesgos_abiertos": 0,
    "contratos_cerrados": 3,
    "lecciones_aprendidas": 8
}

for elemento, valor in proyecto.items():
    print(f"{elemento}: {valor}")
```

## 10. Cierre de recursos

Cuando el proyecto termina, los recursos asignados pueden ser liberados.

Esto puede incluir:

- Personal.
- Equipos.
- Licencias.
- Ambientes de trabajo.
- Servicios contratados.
- Presupuesto restante.

El cierre debe dejar claro qué recursos continúan, cuáles se liberan y qué obligaciones quedan pendientes.

## 11. Informe final

Un informe final puede incluir:

- Objetivo del proyecto.
- Alcance realizado.
- Resultados.
- Cronograma final.
- Costos finales.
- Calidad.
- Riesgos relevantes.
- Cambios realizados.
- Problemas encontrados.
- Lecciones aprendidas.
- Estado de los entregables.

El objetivo es dejar una visión clara del proyecto y de sus resultados.

## 14. Conceptos para recordar

| Concepto | Significado |
|---|---|
| Integración | Coordinación de las diferentes partes del proyecto |
| Cambio | Modificación que puede afectar el plan o resultado |
| Control integrado | Análisis de los efectos de un cambio antes de aplicarlo |
| Entregable | Resultado que debe ser producido y entregado |
| Aceptación | Confirmación de que el resultado cumple lo acordado |
| Cierre | Finalización formal del proyecto |
| Lección aprendida | Conocimiento obtenido durante el proyecto |
| Informe final | Documento que resume resultados y situación final |

## 15. Importancia para Ingeniería de Proyectos

La integración permite entender que todas las decisiones del proyecto están conectadas.

El cierre, por su parte, evita terminar simplemente cuando se entrega el producto. Es necesario completar las actividades administrativas, documentales y de conocimiento.

La secuencia puede resumirse:

**Coordinar → Ejecutar → Controlar → Entregar → Aceptar → Documentar → Cerrar → Aprender**

## 16. Conclusión

La gestión de la integración permite coordinar las diferentes áreas del proyecto y mantenerlas alineadas con los objetivos.

El cierre permite confirmar que el trabajo está terminado, formalizar la entrega, cerrar actividades y conservar las lecciones aprendidas.

Un proyecto no termina solamente cuando se entrega el producto; termina cuando se han completado y documentado correctamente las actividades de cierre.
