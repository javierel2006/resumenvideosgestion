# Dominando el Cronograma - Javier Moreno

## Análisis y resumen del video 5

**Asignatura:** Ingeniería de Proyectos

### Propósito
Comprender cómo se construye un cronograma a partir de las actividades del proyecto, cómo se relacionan entre sí y cómo se estiman y ajustan sus tiempos.

## 1. Resumen del video

El video explica cómo pasar de la organización del trabajo del proyecto a un **cronograma**.

Después de definir el alcance y construir la EDT, es necesario identificar las actividades, establecer el orden en que deben realizarse, estimar su esfuerzo y duración y ajustar el cronograma.

El video presenta seis temas principales:

1. **Dominio de programación.**
2. **Definir actividades.**
3. **Secuencias y lógica.**
4. **Adelantos y retrasos.**
5. **Tutorial aplicado a software.**
6. **Entregables clave.**

La secuencia general presentada es:

**Definir → Secuenciar → Estimar → Ajustar**

El cronograma se construye entendiendo no solamente cuánto dura cada actividad, sino también **qué actividad depende de cuál y qué relación existe entre ellas**.

## 2. Dominio de programación

El video presenta la programación como una parte fundamental de la gestión del tiempo del proyecto.

Programar significa organizar las actividades considerando:

- Qué se debe hacer.
- En qué orden.
- Qué actividades dependen de otras.
- Cuánto esfuerzo requieren.
- Cuánto tiempo pueden tomar.
- Qué ajustes son necesarios.

El cronograma permite convertir una lista de actividades en una secuencia organizada en el tiempo.

## 3. Definir actividades

El primer paso es **identificar y documentar acciones específicas**.

La EDT muestra el trabajo y sus componentes, pero para construir el cronograma necesitamos llegar a actividades que puedan ser organizadas temporalmente.

### Ejemplo de software

Un proyecto puede tener:

**Desarrollo de sistema**
- Diseño del sistema.
- Codificación.
- Pruebas unitarias.
- Despliegue.

Estas actividades pueden posteriormente ubicarse en una secuencia temporal.

El video utiliza precisamente un ejemplo de software donde aparecen:

**Diseño → Codificación → Pruebas unitarias → Despliegue**

## 4. Secuenciar actividades

Después de identificar las actividades hay que determinar su **orden lógico**.

No todas las actividades pueden comenzar al mismo tiempo.

Por ejemplo:

**Diseño**
↓
**Codificación**
↓
**Pruebas**
↓
**Despliegue**

En este caso, las pruebas dependen de que exista código que pueda probarse.

Secuenciar significa entonces identificar las relaciones entre actividades y determinar qué actividad es predecesora y cuál es sucesora.

## 5. Método de diagramación de precedencia

El video presenta las relaciones del **método de diagramación de precedencia**.

Las cuatro relaciones principales son:

### FS — Finish to Start
**Fin a Inicio**

La actividad sucesora comienza cuando termina la predecesora.

**Ejemplo:**

Diseño termina → Codificación comienza.

### SS — Start to Start
**Inicio a Inicio**

La actividad sucesora puede comenzar cuando comienza la predecesora.

**Ejemplo:**

Dos actividades pueden comenzar aproximadamente al mismo tiempo.

### FF — Finish to Finish
**Fin a Fin**

La finalización de una actividad está relacionada con la finalización de otra.

### SF — Start to Finish
**Inicio a Fin**

La actividad sucesora depende del inicio de la predecesora para poder finalizar.

Esta última relación es menos habitual en proyectos comunes, pero forma parte de las relaciones posibles del método de precedencia.

## 6. Adelantos y retrasos

El video diferencia dos conceptos:

### Adelanto

Es el **tiempo en que una actividad sucesora puede adelantarse** respecto de la relación normal con su predecesora.

Permite comenzar una actividad antes de que termine completamente la actividad anterior, cuando la lógica del trabajo lo permite.

### Retraso

Es el **tiempo de espera entre una sucesora y su predecesora**.

Por ejemplo, después de terminar una actividad puede ser necesario esperar determinado tiempo antes de comenzar la siguiente.

### Resumen

**Adelanto = permite anticipar**

**Retraso = agrega tiempo de espera**

## 7. Estimar esfuerzo y duración

Después de definir y secuenciar las actividades, el video presenta la estimación.

Es importante distinguir:

### Esfuerzo
Cantidad de trabajo necesaria para realizar una actividad.

### Duración
Tiempo transcurrido necesario para completar la actividad.

No siempre significan lo mismo.

Por ejemplo, una actividad puede requerir cierta cantidad de trabajo, pero su duración puede depender de cuántas personas participan, de la disponibilidad de recursos o de las dependencias con otras actividades.

## 8. Ajustar el cronograma

El cronograma no debe considerarse completamente terminado en el primer intento.

El video presenta la idea de **revisar y modificar el borrador**.

Los ajustes pueden ser necesarios por:

- Cambios en las dependencias.
- Disponibilidad de recursos.
- Duraciones estimadas.
- Adelantos.
- Retrasos.
- Restricciones.
- Necesidades del proyecto.

Por eso el cronograma debe revisarse antes de utilizarlo como referencia para el seguimiento.

## 9. Ejemplo aplicado al desarrollo de software

El video utiliza un ejemplo sencillo:

| ID | Actividad | Predecesora | Relación |
|---|---|---|---|
| 1 | Diseño | Ninguna | N/A |
| 2 | Código | 1 | FS |
| 3 | Pruebas | 2 | FS |

La lógica es:

**Diseño → Código → Pruebas**

La primera actividad no tiene predecesora.

La actividad de código depende del diseño mediante una relación **FS**.

Las pruebas dependen del código mediante otra relación **FS**.

El ejemplo muestra cómo una lista de actividades se convierte en una estructura lógica que posteriormente puede ubicarse en el tiempo.

## 10. El cronograma como una cadena lógica

Una idea importante del video es que el cronograma no consiste solamente en colocar fechas en un calendario.

Existe una **cadena lógica de actividades**.

Por ejemplo:

**Diseño**
→ **Código**
→ **Pruebas**
→ **Producto terminado**

Si una actividad se retrasa y las actividades posteriores dependen de ella, el retraso puede afectar el resto de la secuencia.

Por eso es importante comprender las relaciones entre actividades antes de establecer las fechas definitivas.

## 11. Entregables clave

El video termina relacionando el cronograma con los entregables.

En un proyecto de software pueden existir entregables como:

- Diseño del sistema.
- Código desarrollado.
- Pruebas realizadas.
- Sistema desplegado.
- Producto terminado.

El cronograma ayuda a organizar cuándo se espera producir cada uno.

De esta manera se conecta lo estudiado anteriormente:

**EDT → Actividades → Secuencia → Duración → Cronograma → Entregables**

## 12. Análisis del video

El principal aprendizaje es que **un cronograma bien construido depende de una buena definición del trabajo**.

No tendría sentido asignar fechas a actividades que todavía no están claramente identificadas.

La progresión correcta es:

1. Identificar las actividades.
2. Definir sus relaciones.
3. Estimar esfuerzo y duración.
4. Revisar las dependencias.
5. Aplicar adelantos o retrasos cuando corresponda.
6. Ajustar el cronograma.
7. Relacionarlo con los entregables.

Esto permite tener una visión más realista del desarrollo temporal del proyecto.

## 15. Conclusión

El video muestra cómo transformar el trabajo definido en la EDT en un cronograma organizado.

La clave está en comprender que las actividades tienen **secuencias, relaciones, esfuerzo y duración**.

La construcción del cronograma puede resumirse en:

**Definir → Secuenciar → Estimar → Ajustar**

En un proyecto de software, esto permite pasar de una lista general de trabajo a una secuencia concreta como:

**Diseño → Codificación → Pruebas → Despliegue**

Esta base será necesaria para el siguiente video, donde se profundizará en la planificación y organización del tiempo del proyecto.

## 16. Actividad de análisis

Para un proyecto sencillo de software, identifica cinco actividades.

Después establece:

1. Nombre de la actividad.
2. Actividad predecesora.
3. Relación entre ellas.
4. Duración estimada.
5. Si existe algún adelanto o retraso.

Finalmente organiza las actividades en el siguiente orden:

**Actividad → Predecesora → Relación → Duración → Entregable**

El objetivo es practicar la construcción lógica de un cronograma antes de pasar a herramientas más avanzadas.
