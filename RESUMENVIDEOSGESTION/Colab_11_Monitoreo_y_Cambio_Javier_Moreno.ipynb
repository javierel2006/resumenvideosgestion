# Monitoreo y Cambio - Javier Moreno

## Análisis y resumen del video 11

**Asignatura:** Ingeniería de Proyectos

### Propósito
Comprender cómo se realiza el seguimiento del proyecto, cómo se comparan los resultados reales con el plan y cómo se gestionan los cambios durante la ejecución.

## 1. Resumen del video

El video explica que la planificación del proyecto establece una referencia, pero durante la ejecución pueden aparecer diferencias, problemas y nuevas necesidades.

Por esta razón es necesario **monitorear el proyecto** y controlar los cambios.

El seguimiento permite comparar:

**Lo planificado ↔ Lo realizado**

A partir de esa comparación se pueden identificar:

- Variaciones.
- Retrasos.
- Diferencias de costo.
- Problemas de desempeño.
- Cambios necesarios.
- Riesgos que requieren atención.

El video relaciona el monitoreo con la toma de decisiones y muestra que un cambio debe analizarse antes de incorporarse al proyecto.

## 2. ¿Qué es monitorear?

Monitorear significa observar y recopilar información sobre el comportamiento del proyecto.

El gerente necesita conocer, entre otras cosas:

- Qué trabajo se ha completado.
- Qué trabajo está pendiente.
- Cuánto tiempo se ha utilizado.
- Cuánto dinero se ha gastado.
- Qué problemas han aparecido.
- Si existen desviaciones frente al plan.

La idea sencilla es:

**Planificar → Ejecutar → Medir → Comparar → Actuar**

Monitorear no significa solamente mirar si una actividad terminó. Significa obtener información que permita tomar decisiones.

## 3. Planificado frente a real

Una comparación básica del seguimiento es:

| Elemento | Planificado | Real |
|---|---|---|
| Tiempo | Lo que se esperaba | Lo que realmente ocurrió |
| Costo | Lo presupuestado | Lo gastado |
| Trabajo | Lo que debía completarse | Lo que se completó |
| Entregables | Lo esperado | Lo entregado |

Las diferencias entre ambas columnas permiten detectar desviaciones.

### Ejemplo

Si se esperaba terminar una actividad en 5 días y realmente tomó 7 días:

**Plan = 5 días**

**Real = 7 días**

Existe una diferencia de:

**2 días de retraso**

Lo importante no es solamente identificar la diferencia, sino analizar su causa y su impacto.

## 4. Desviaciones

Una desviación aparece cuando el comportamiento real se separa de lo establecido en el plan.

Puede existir desviación de:

- Tiempo.
- Costo.
- Alcance.
- Recursos.
- Calidad.

### Ejemplo de tiempo

**Planificado: 10 días**

**Real: 12 días**

**Desviación: +2 días**

### Ejemplo de costo

**Planificado: $10.000.000**

**Real: $11.500.000**

**Desviación: +$1.500.000**

Una desviación no significa automáticamente que el proyecto haya fracasado. Primero debe analizarse qué la produjo y qué consecuencias puede tener.

## 5. El ciclo de control

El monitoreo puede entenderse como un ciclo:

**1. Planificar**
↓
**2. Ejecutar**
↓
**3. Medir**
↓
**4. Comparar**
↓
**5. Analizar**
↓
**6. Tomar acción**
↓
**7. Actualizar cuando corresponda**

Este ciclo se repite durante la ejecución del proyecto.

La ventaja es que los problemas pueden detectarse antes de que se conviertan en problemas mayores.

## 6. ¿Qué es un cambio?

Un cambio es una modificación respecto de lo que había sido establecido.

Puede afectar:

- Alcance.
- Cronograma.
- Costos.
- Recursos.
- Calidad.
- Requisitos.
- Entregables.

### Ejemplo

El cliente solicita agregar una nueva función al sistema cuando el desarrollo ya está avanzado.

No basta con responder inmediatamente:

**“Sí, la agregamos.”**

Primero es necesario analizar:

- Qué trabajo adicional implica.
- Cuánto tiempo requiere.
- Cuánto puede costar.
- Qué recursos necesita.
- Qué otras actividades puede afectar.
- Si cambia el alcance aprobado.

## 7. Control integrado de cambios

El cambio debe pasar por un proceso de análisis y decisión.

Una secuencia sencilla es:

**Solicitud de cambio**
↓
**Analizar impacto**
↓
**Evaluar alternativas**
↓
**Decidir**
↓
**Aprobar o rechazar**
↓
**Actualizar el plan si se aprueba**
↓
**Comunicar**
↓
**Implementar**
↓
**Verificar**

La idea fundamental es evitar que los cambios entren al proyecto sin conocer sus consecuencias.

## 8. Impacto de un cambio

Un cambio aparentemente pequeño puede afectar varias áreas.

Por ejemplo:

**Agregar una funcionalidad**

puede producir:

**Más trabajo**
→ **Más horas**
→ **Más costo**
→ **Mayor duración**
→ **Nuevos recursos**
→ **Nuevas pruebas**

Por eso el análisis debe ser integral.

No debe evaluarse únicamente el elemento que cambia, sino también las consecuencias sobre el resto del proyecto.

## 9. Línea base y cambios

El video se relaciona directamente con el tema anterior de **Estimación y Línea Base**.

La línea base representa una referencia aprobada.

Cuando aparece un cambio aprobado que modifica esa referencia, es necesario actualizar la información correspondiente de acuerdo con el proceso de control de cambios.

La relación puede resumirse:

**Línea base**
↓
**Ejecución**
↓
**Monitoreo**
↓
**Desviación o solicitud**
↓
**Análisis**
↓
**Cambio aprobado**
↓
**Actualización correspondiente**

Esto permite mantener una diferencia clara entre lo que estaba aprobado originalmente y las modificaciones aprobadas posteriormente.

## 10. Seguimiento de alcance, tiempo y costo

El monitoreo debe observar varias dimensiones simultáneamente.

### Alcance
¿Se está realizando el trabajo aprobado?

### Tiempo
¿El proyecto avanza según el cronograma?

### Costo
¿El gasto se mantiene dentro de lo esperado?

### Calidad
¿Los entregables cumplen los requisitos?

### Recursos
¿Las personas y recursos están disponibles y se utilizan adecuadamente?

Estas dimensiones están relacionadas.

Un cambio en una puede producir efectos en las demás.

## 11. Análisis del video

La principal enseñanza es que **un proyecto no se controla únicamente al final**.

El control debe realizarse durante la ejecución para detectar diferencias y tomar acciones oportunas.

También se observa que los cambios son una parte normal de los proyectos, especialmente cuando existen necesidades nuevas o información que no estaba disponible al principio.

El problema no es que exista un cambio.

El verdadero desafío es **comprender su impacto y gestionarlo de manera ordenada**.

Por eso el monitoreo y el control de cambios están estrechamente relacionados.

## 13. Ejemplo sencillo

Supongamos que un proyecto planifica:

**Desarrollo del módulo de usuarios: 10 días**

Durante la ejecución se observa:

**Duración real: 13 días**

La diferencia es:

**13 - 10 = 3 días**

El equipo debe investigar:

1. ¿Por qué ocurrió el retraso?
2. ¿Afecta actividades posteriores?
3. ¿Afecta la ruta crítica?
4. ¿Aumenta el costo?
5. ¿Es necesario realizar alguna acción?
6. ¿Existe un cambio aprobado que explique la diferencia?

El objetivo del monitoreo es convertir esta información en una decisión adecuada.

## 14. Ejemplo de solicitud de cambio

### Situación

El cliente solicita agregar un módulo de reportes.

### Análisis inicial

**Alcance:** aumenta.

**Tiempo:** se requieren nuevas actividades.

**Costo:** puede aumentar por horas de trabajo adicionales.

**Recursos:** puede ser necesario asignar desarrolladores y testers.

**Calidad:** será necesario realizar nuevas pruebas.

### Decisión

Antes de incorporar el módulo, la solicitud debe analizarse y seguir el proceso de aprobación establecido.

Esto evita comprometer al equipo con un trabajo que no estaba incluido en el plan original sin conocer sus consecuencias.

## 15. Conceptos para recordar

| Concepto | Explicación sencilla |
|---|---|
| Monitoreo | Observación y recopilación de información del proyecto |
| Control | Comparación y acciones para mantener el proyecto bajo control |
| Planificado | Lo que se había establecido |
| Real | Lo que realmente ocurrió |
| Desviación | Diferencia entre el plan y el resultado |
| Cambio | Modificación respecto de lo aprobado |
| Impacto | Consecuencias que puede producir un cambio |
| Solicitud de cambio | Propuesta formal para modificar algún elemento |
| Línea base | Referencia aprobada para realizar comparaciones |
| Control integrado de cambios | Proceso para analizar, decidir y gestionar cambios |

## 16. Importancia para Ingeniería de Proyectos

El monitoreo permite al gerente conocer el estado real del proyecto.

Esto ayuda a:

- Detectar retrasos.
- Identificar sobrecostos.
- Controlar el alcance.
- Revisar recursos.
- Detectar problemas.
- Analizar cambios.
- Tomar decisiones oportunas.

En un proyecto de software es especialmente importante porque los requisitos y las condiciones pueden evolucionar durante el desarrollo.

Una gestión ordenada permite aceptar cambios cuando corresponda sin perder el control del proyecto.

## 17. Conclusión

El video muestra que la gestión del proyecto continúa después de la planificación.

Durante la ejecución se debe:

**Medir → Comparar → Analizar → Actuar**

Cuando aparece una diferencia, es necesario comprender su causa.

Cuando aparece una solicitud de cambio, debe analizarse su impacto antes de aprobarla.

La idea principal puede resumirse:

> Monitorear permite conocer qué está ocurriendo; controlar cambios permite decidir de manera ordenada qué debe modificarse.
