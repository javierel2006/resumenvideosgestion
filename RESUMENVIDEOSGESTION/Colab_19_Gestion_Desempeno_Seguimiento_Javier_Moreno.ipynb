### Propósito
Comprender cómo hacer seguimiento al proyecto, comparar lo planificado con lo realizado y utilizar la información obtenida para tomar decisiones.

## 1. Resumen del tema

El seguimiento permite conocer cómo está avanzando realmente un proyecto.

No basta con saber que el equipo está trabajando. Es necesario comprobar:

- Qué actividades se han completado.
- Qué actividades están pendientes.
- Cuánto se ha gastado.
- Qué resultados se han obtenido.
- Qué problemas existen.
- Qué cambios han aparecido.
- Qué riesgos requieren atención.

El seguimiento convierte los datos del proyecto en información útil para decidir.

## 2. Planificado frente a real

Una de las comparaciones más importantes es:

**Planificado:** lo que se esperaba realizar.

**Real:** lo que realmente se realizó.

Ejemplo:

Se esperaba completar 20 tareas durante el mes.

Al finalizar el mes se completaron 16.

Existe una diferencia de 4 tareas que debe analizarse.

La diferencia por sí sola no explica la causa. Es necesario investigar qué ocurrió.

## 3. Indicadores de desempeño

Los indicadores ayudan a resumir el estado del proyecto.

Pueden medir:

- Avance.
- Tiempo.
- Costo.
- Calidad.
- Riesgos.
- Incidencias.
- Cumplimiento de requisitos.

Un indicador debe servir para responder una pregunta concreta y facilitar el seguimiento.

## 4. Porcentaje de avance

Una forma sencilla de calcular el avance es:

**Avance (%) = trabajo completado / trabajo total × 100**

Ejemplo:

Si un proyecto tiene 50 tareas y se han completado 35:

**Avance = 35 / 50 × 100 = 70%**

Esto indica que se ha completado el 70% de las tareas consideradas en el ejemplo.

## 5. Desviaciones

Una desviación aparece cuando el resultado real no coincide con la referencia planificada.

Puede existir desviación en:

- Tiempo.
- Costo.
- Alcance.
- Calidad.
- Recursos.

Ejemplo:

**Plan:** terminar una funcionalidad el día 10.

**Real:** se terminó el día 15.

Existe una desviación de cinco días.

## 6. Análisis de causas

Cuando se detecta una desviación, el siguiente paso es buscar la causa.

Algunas causas pueden ser:

- Estimación incorrecta.
- Falta de recursos.
- Cambios de requisitos.
- Dependencias externas.
- Problemas técnicos.
- Retrabajo.
- Problemas de comunicación.
- Riesgos que se materializaron.

La respuesta debe dirigirse a la causa y no solamente al síntoma.

## 7. Acciones correctivas

Cuando el desempeño se desvía del plan pueden definirse acciones correctivas.

Ejemplos:

- Reorganizar actividades.
- Asignar recursos adicionales.
- Resolver una dependencia.
- Ajustar prioridades.
- Corregir un problema técnico.
- Revisar una estimación.
- Gestionar un cambio formal.

La acción debe estar relacionada con la causa identificada.

## 8. Acciones preventivas

También pueden realizarse acciones antes de que aparezca un problema.

Ejemplos:

- Realizar pruebas tempranas.
- Revisar requisitos.
- Capacitar al equipo.
- Verificar dependencias.
- Preparar recursos alternativos.
- Actualizar el registro de riesgos.

La prevención ayuda a disminuir la posibilidad de desviaciones futuras.

```python
# Cálculo sencillo del avance de un proyecto

tareas_totales = 50
tareas_completadas = 35

avance = (tareas_completadas / tareas_totales) * 100

print(f"Avance del proyecto: {avance:.1f}%")
```

## 9. Tablero de seguimiento

Este tipo de información permite detectar rápidamente dónde se necesita atención.

## 10. Reuniones de seguimiento

2. ¿Qué está pendiente?
3. ¿Qué problemas existen?

## 11. Seguimiento en proyectos de software

En software se pueden revisar:

## 12. Relación con el Valor Ganado

El seguimiento también puede utilizar indicadores del Valor Ganado:

**PV** → valor planificado.

**EV** → valor ganado.

**AC** → costo real.

Con estos datos se pueden calcular:

**SV = EV - PV**

**CV = EV - AC**

**SPI = EV / PV**

**CPI = EV / AC**

Así, el seguimiento permite relacionar el avance con el tiempo y el costo.

## 13. Seguimiento y toma de decisiones

El objetivo del seguimiento no es producir informes por sí mismos.

La información debe ayudar a decidir.

Por ejemplo:

**Dato:** una actividad tiene retraso.

↓

**Análisis:** la causa es una dependencia externa.

↓

**Decisión:** reorganizar las actividades mientras se resuelve la dependencia.

↓

**Seguimiento:** comprobar si la acción corrigió la desviación.

## 14. Relación con los temas anteriores

El seguimiento reúne información de diferentes áreas:

**Cronograma**
→ muestra fechas y actividades.

**Costos**
→ muestra utilización del presupuesto.

**Riesgos**
→ muestra amenazas y oportunidades.

**Calidad**
→ muestra defectos y resultados de pruebas.

**Interesados**
→ permite comunicar el estado.

**Cambios**
→ permite controlar modificaciones.

**Valor Ganado**
→ permite medir desempeño de tiempo y costo.

Por eso el seguimiento es una actividad transversal del proyecto.

## 17. Importancia para Ingeniería de Proyectos

El seguimiento permite detectar problemas mientras todavía existe tiempo para actuar.

La secuencia puede resumirse:

**Planificar → Ejecutar → Medir → Comparar → Analizar → Actuar → Volver a medir**

Esto convierte el seguimiento en un proceso continuo y no en una actividad realizada únicamente al final.

Un buen seguimiento debe producir información clara, decisiones concretas y responsables para las acciones acordadas.
