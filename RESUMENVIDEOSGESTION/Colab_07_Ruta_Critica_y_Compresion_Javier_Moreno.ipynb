# Ruta Crítica y Compresión - Javier Moreno

## Análisis y resumen del video 7

**Asignatura:** Ingeniería de Proyectos

### Propósito

## 1. Resumen del video

El video explica cómo determinar qué actividades tienen mayor influencia sobre la duración total del proyecto.

El concepto central es la **ruta crítica**, formada por la secuencia de actividades que determina la duración mínima del proyecto.

El video muestra:

- Cálculo hacia adelante.
- Cálculo hacia atrás.
- Inicio y finalización tempranos.
- Inicio y finalización tardíos.
- Holgura.
- Identificación de actividades críticas.
- Fecha límite o restricción del proyecto.
- Técnicas de compresión.
- Diferencia entre **Fast Tracking** y **Crashing**.

La idea puede resumirse así:

**Red de actividades → Cálculo hacia adelante → Cálculo hacia atrás → Holguras → Ruta crítica → Análisis de compresión**

## 2. ¿Qué es la ruta crítica?

La **ruta crítica** es la secuencia de actividades que determina la duración total del proyecto.

Las actividades que forman parte de esta ruta normalmente tienen **holgura total igual a cero**.

Esto significa que un retraso en una actividad crítica puede afectar la fecha final del proyecto, si no existen otros mecanismos de recuperación.

### Ejemplo sencillo

Si tenemos:

**A → B → C → D**

y esta es la secuencia que determina la duración total, entonces A-B-C-D forma la ruta crítica.

No todas las actividades del proyecto necesariamente son críticas. Algunas pueden tener cierto margen de tiempo sin afectar la fecha final.

## 3. Cálculo hacia adelante

El video presenta el cálculo **hacia adelante**.

Su objetivo es determinar:

- **Inicio temprano (ES).**
- **Finalización temprana (EF).**

La pregunta que responde es:

> ¿Cuál es la fecha más temprana en que una actividad puede comenzar y terminar?

Una relación básica es:

**EF = ES + Duración**

Para una actividad que depende de otra, su inicio temprano se determina considerando la finalización temprana de sus predecesoras.

### Idea sencilla

**Hacia adelante = calcular cuándo puede comenzar y terminar lo antes posible.**

## 4. Cálculo hacia atrás

Después se realiza el cálculo **hacia atrás**.

Su objetivo es determinar:

- **Inicio tardío (LS).**
- **Finalización tardía (LF).**

La pregunta es:

> ¿Hasta cuándo puede retrasarse una actividad sin afectar la fecha final establecida del proyecto?

Una relación básica es:

**LS = LF - Duración**

El cálculo parte de la fecha final del proyecto y retrocede por la red de actividades.

### Idea sencilla

**Hacia atrás = calcular hasta cuándo puede comenzar o terminar una actividad sin retrasar el proyecto.**

## 5. Holgura

La **holgura** indica cuánto puede retrasarse una actividad sin afectar determinada fecha del proyecto.

Una forma sencilla de calcular la holgura total es:

**Holgura = LS - ES**

También puede calcularse como:

**Holgura = LF - EF**

### Interpretación

- **Holgura = 0:** actividad crítica.
- **Holgura mayor que 0:** existe margen de tiempo.

Por eso la holgura es fundamental para identificar la ruta crítica.

## 6. Ejemplo de cálculo

Supongamos una actividad con:

- Inicio temprano (ES) = 4.
- Finalización temprana (EF) = 8.
- Inicio tardío (LS) = 4.
- Finalización tardía (LF) = 8.

Entonces:

**Holgura = LS - ES**

**Holgura = 4 - 4 = 0**

La actividad es crítica.

Ahora supongamos:

- ES = 4.
- LS = 6.

Entonces:

**Holgura = 6 - 4 = 2**

Existe una holgura de 2 unidades de tiempo.

## 7. Fecha límite del proyecto

El video presenta un escenario en el que el cálculo inicial puede superar una **fecha límite inamovible de lanzamiento al mercado**.

Esto significa que el proyecto necesita terminar antes de una fecha establecida.

Cuando el cronograma calculado no cumple la fecha límite, el equipo debe estudiar alternativas.

Una de ellas es analizar las actividades de la ruta crítica y determinar si es posible reducir la duración del proyecto mediante técnicas de compresión.

## 8. Compresión del cronograma

La **compresión del cronograma** busca reducir la duración del proyecto sin cambiar su alcance.

El video presenta principalmente dos técnicas:

### Fast Tracking — Intensificación / ejecución en paralelo

Consiste en realizar en paralelo actividades que normalmente se harían de manera secuencial, cuando la relación entre ellas lo permite.

**Ventaja:** puede reducir tiempo.

**Riesgo:** aumenta la posibilidad de retrabajo o problemas por realizar actividades superpuestas.

### Crashing — Aceleración / intensificación de recursos

Consiste en agregar recursos o aumentar la capacidad utilizada para reducir la duración de determinadas actividades.

**Ventaja:** puede reducir la duración.

**Desventaja:** normalmente implica mayor costo.

El video destaca especialmente la relación entre **tiempo, riesgo y costo**.

## 9. Diferencia entre Fast Tracking y Crashing

| Técnica | Cómo reduce el tiempo | Impacto principal |
|---|---|---|
| Fast Tracking | Superpone actividades | Mayor riesgo |
| Crashing | Agrega recursos o capacidad | Mayor costo |

### Ejemplo

**Fast Tracking:**

Comenzar parte de las pruebas mientras todavía se termina una parte del desarrollo, si técnicamente es posible.

**Crashing:**

Asignar más desarrolladores a una actividad crítica para intentar reducir su duración.

La elección depende de las condiciones reales del proyecto.

## 10. Análisis del video

El principal aprendizaje es que no todas las actividades merecen el mismo nivel de atención cuando el objetivo es reducir la duración total.

Primero es necesario conocer:

**¿Cuál es la ruta crítica?**

Después se analiza:

**¿Qué actividades críticas pueden comprimirse?**

Esto evita invertir tiempo o recursos en actividades que no reducirían la duración total del proyecto.

También es importante entender que reducir tiempo puede generar consecuencias:

**Menor tiempo → posible mayor costo o mayor riesgo**

Por eso la compresión debe analizarse antes de aplicarse.

## 12. Ejemplo sencillo de red

Supongamos:

| Actividad | Duración | Predecesora |
|---|---:|---|
| A | 3 días | — |
| B | 4 días | A |
| C | 2 días | A |
| D | 5 días | B |
| E | 3 días | C |
| F | 2 días | D y E |

Existen dos caminos principales:

**A → B → D → F**

Duración:

**3 + 4 + 5 + 2 = 14 días**

**A → C → E → F**

Duración:

**3 + 2 + 3 + 2 = 10 días**

En este ejemplo, el camino de 14 días determina la duración del proyecto y constituye la ruta crítica, bajo estos supuestos.

## 13. Importancia para Ingeniería de Proyectos

La ruta crítica permite al gerente de proyectos identificar dónde un retraso puede tener mayor impacto sobre la fecha final.

También ayuda a:

- Priorizar seguimiento.
- Analizar fechas límite.
- Estudiar alternativas de reducción de tiempo.
- Evaluar recursos adicionales.
- Analizar riesgos de programación.
- Tomar decisiones sobre el cronograma.

La ruta crítica no significa que las demás actividades sean irrelevantes. Significa que determinadas actividades tienen una relación directa con la duración total del proyecto.

## 15. Conclusión

El video enseña que conocer el cronograma no es suficiente. También es necesario identificar qué actividades determinan la duración total.

Para ello se utilizan los cálculos hacia adelante y hacia atrás y se obtiene la holgura de cada actividad.

La lógica puede resumirse:

**Red de actividades → Fechas tempranas → Fechas tardías → Holgura → Ruta crítica**

Cuando existe una fecha límite que no puede modificarse, pueden analizarse técnicas de compresión como:

**Fast Tracking** y **Crashing**.

La primera puede aumentar el riesgo y la segunda puede aumentar el costo. Por eso la decisión debe basarse en las condiciones del proyecto.
