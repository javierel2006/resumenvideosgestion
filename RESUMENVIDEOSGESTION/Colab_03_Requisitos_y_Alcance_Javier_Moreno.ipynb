# Requisitos y Alcance - Javier Moreno

## Análisis y resumen del video 3

**Asignatura:** Ingeniería de Proyectos

### Propósito
Comprender cómo se identifican los requisitos y cómo se define el alcance de un proyecto, diferenciando claramente el alcance del producto y el alcance del proyecto.

## 1. Resumen del video

El video explica dos elementos fundamentales para iniciar y organizar un proyecto: **los requisitos y el alcance**.

Se presenta el requisito como una capacidad necesaria para satisfacer una necesidad empresarial. Por esta razón, los requisitos deben estar relacionados con una necesidad real y no simplemente con ideas aisladas.

También se diferencia entre:

- **Alcance del producto:** características y funciones que tendrá el producto.
- **Alcance del proyecto:** trabajo que debe realizarse para entregar ese producto.

El video muestra además cómo una idea general puede desglosarse progresivamente hasta llegar a elementos concretos, incluso hasta las tareas que permiten realizar el trabajo.

La idea general puede resumirse así:

**Necesidad → Requisitos → Producto → Trabajo del proyecto → Tareas**

## 2. ¿Qué es un requisito?

El video define un requisito como una **capacidad necesaria para satisfacer una necesidad empresarial**.

Esta definición permite entender que un requisito no es simplemente una tarea.

### Ejemplo

Si una empresa necesita mejorar el control de sus ventas, una necesidad podría ser:

**Necesidad:** mejorar el control de las ventas.

A partir de ella pueden aparecer requisitos como:

- Registrar ventas.
- Consultar ventas.
- Generar reportes.
- Controlar usuarios.
- Proteger la información.

Cada requisito representa una capacidad que el sistema necesita para responder a la necesidad planteada.

## 3. Requisitos y necesidades

Una buena definición de requisitos debe comenzar por comprender la necesidad.

La relación puede representarse de esta manera:

**Necesidad empresarial**
↓
**Requisito**
↓
**Característica o capacidad**
↓
**Resultado esperado**

Esto es importante porque ayuda a evitar que el proyecto se llene de funciones que no aportan al objetivo principal.

### Pregunta que debe hacerse el equipo

**¿Este requisito realmente ayuda a satisfacer la necesidad que originó el proyecto?**

Si la respuesta es no, debe revisarse su importancia antes de incorporarlo al alcance.

## 4. Alcance del producto

El video diferencia el **alcance del producto** del alcance del proyecto.

El alcance del producto corresponde a las:

- Características.
- Funciones.
- Capacidades.
- Propiedades.

que tendrá el producto o resultado final.

### Ejemplo

Para una aplicación de ventas, el alcance del producto podría incluir:

- Inicio de sesión.
- Registro de clientes.
- Registro de productos.
- Registro de ventas.
- Consulta de información.
- Generación de reportes.

Estas son características que tendrá el producto entregado.

## 5. Alcance del proyecto

El **alcance del proyecto** corresponde al trabajo que debe realizarse para entregar el producto, servicio o resultado.

Por ejemplo, para crear una aplicación de ventas el proyecto puede incluir:

- Levantamiento de requisitos.
- Diseño.
- Programación.
- Pruebas.
- Corrección de errores.
- Documentación.
- Implementación.
- Capacitación.

Por lo tanto:

**Alcance del producto = qué tendrá el producto**

**Alcance del proyecto = qué trabajo debemos realizar para entregarlo**

## 6. Diferencia entre producto y proyecto

| Elemento | Pregunta principal | Ejemplo |
|---|---|---|
| Alcance del producto | ¿Qué tendrá el producto? | Sistema con login y reportes |
| Alcance del proyecto | ¿Qué debemos hacer para entregarlo? | Analizar, diseñar, programar y probar |

Esta diferencia es fundamental porque evita confundir las características del resultado con el trabajo necesario para producirlo.

## 7. Desglose progresivo

Uno de los elementos visuales importantes del video muestra el **desglose desde la visión hasta las tareas**.

La idea consiste en comenzar con una visión general y avanzar hacia elementos cada vez más concretos.

Una secuencia sencilla sería:

**Visión**
↓
**Objetivo**
↓
**Beneficio**
↓
**Producto**
↓
**Características**
↓
**Trabajo**
↓
**Tareas**

Este desglose permite convertir una idea general en un conjunto organizado de elementos que pueden ser gestionados por el equipo.

## 8. Relación entre rol, objetivo y beneficio

El video presenta una representación sencilla con tres elementos:

**[Rol] + [Objetivo] + [Beneficio]**

Esta estructura ayuda a pensar los requisitos desde la perspectiva de quien necesita o utilizará el resultado.

Por ejemplo:

> Como administrador, quiero consultar las ventas para conocer el comportamiento de la empresa.

Aquí:

- **Rol:** administrador.
- **Objetivo:** consultar las ventas.
- **Beneficio:** conocer el comportamiento de la empresa.

Este tipo de formulación permite relacionar lo que se solicita con la persona que lo necesita y con el beneficio esperado.

## 9. Análisis del video

El punto central del video es que **definir bien los requisitos y el alcance permite saber qué se debe construir y qué trabajo debe realizarse**.

Si los requisitos no están claros, pueden aparecer problemas como:

- Cambios constantes.
- Trabajo innecesario.
- Confusión entre los interesados y el equipo.
- Aumento de tiempo.
- Aumento de costos.
- Entregables que no responden a la necesidad.

Por eso, el alcance funciona como una referencia para determinar qué está incluido y qué no está incluido en el proyecto.

## 10. Relación con Ingeniería de Proyectos

Este tema conecta directamente con la planificación del proyecto.

Antes de elaborar un cronograma detallado, el equipo necesita saber **qué debe entregar y qué trabajo será necesario para lograrlo**.

La progresión es:

**Necesidad**
→ **Requisitos**
→ **Alcance**
→ **Entregables**
→ **Trabajo**
→ **Actividades**
→ **Cronograma**

Esto prepara el camino para el siguiente tema de la secuencia: la **EDT y la EDV**, donde el trabajo puede organizarse y descomponerse de forma estructurada.

## 11. Conceptos para recordar

| Concepto | Explicación sencilla |
|---|---|
| Necesidad | Situación que requiere una solución |
| Requisito | Capacidad necesaria para satisfacer una necesidad |
| Alcance del producto | Características y funciones del resultado |
| Alcance del proyecto | Trabajo necesario para entregar el resultado |
| Entregable | Resultado que debe producir el proyecto |
| Desglose | División progresiva de lo general en elementos concretos |
| Rol | Persona o grupo que necesita o participa en el resultado |
| Beneficio | Resultado útil obtenido al satisfacer la necesidad |

## 12. Conclusión

El video enseña que un proyecto debe comenzar entendiendo claramente la necesidad que se quiere solucionar.

A partir de esa necesidad se identifican los requisitos y se define el alcance.

La diferencia entre **alcance del producto** y **alcance del proyecto** es especialmente importante:

- El producto define las características y funciones que tendrá el resultado.
- El proyecto define el trabajo necesario para conseguirlo.

Una vez definido esto, el trabajo puede descomponerse progresivamente hasta llegar a actividades y tareas concretas.

## 13. Actividad de análisis

Para un proyecto de software, completa la siguiente relación:

**Necesidad:**  
¿Qué problema se quiere solucionar?

**Requisitos:**  
¿Qué capacidades necesita la solución?

**Alcance del producto:**  
¿Qué características tendrá?

**Alcance del proyecto:**  
¿Qué trabajo será necesario realizar?

**Beneficio:**  
¿Qué se espera conseguir con la solución?

La finalidad es comprobar que existe una relación lógica entre la necesidad inicial y el resultado final.

> Los requisitos representan capacidades necesarias para satisfacer necesidades, mientras que el alcance define qué tendrá el producto y qué trabajo debe realizarse para entregarlo.
