## Análisis y resumen del video 6

**Asignatura:** Ingeniería de Proyectos

### Propósito
Comprender cómo realizar estimaciones de esfuerzo y duración utilizando datos, diferentes técnicas de estimación y modelos adecuados al tipo de proyecto.

## 1. Resumen del video

El video plantea una pregunta central:

**¿Cómo podemos predecir el futuro de un proyecto?**

La respuesta presentada se basa en realizar **estimaciones utilizando datos y métodos de cálculo**, en lugar de depender solamente de una opinión o de una aproximación sin fundamento.

Se explican principalmente:

- La diferencia entre **esfuerzo y duración**.
- El uso de datos históricos.
- La definición de parámetros.
- La aplicación de algoritmos.
- Diferentes técnicas de estimación.
- La diferencia entre enfoques predictivos y adaptativos.
- Un ejemplo aplicado a proyectos de software.

La idea principal es que una estimación rigurosa utiliza información disponible para construir una previsión razonable del trabajo futuro.

## 2. Esfuerzo y duración

El video hace una diferenciación fundamental:

### Esfuerzo

Es la cantidad de trabajo necesaria para realizar una actividad.

Puede expresarse, por ejemplo, en **horas de trabajo**.

### Duración

Es el tiempo de calendario que transcurre hasta terminar una actividad.

Puede expresarse en:

- Días.
- Semanas.
- Meses.
- Sprints.

### Ejemplo

Una actividad puede requerir:

**120 horas de esfuerzo**

pero tener una:

**Duración de 3 semanas**

Esto puede ocurrir porque varias personas participan, existen días no laborables o el trabajo se distribuye dentro del calendario.

Por eso **120 horas no significa automáticamente 120 horas de duración del proyecto**.

## 3. Importancia de los datos históricos

El video muestra que una estimación mejora cuando se dispone de información de proyectos anteriores.

Los datos históricos pueden ayudar a conocer:

- Cuánto tiempo tomó un trabajo similar.
- Cuánto esfuerzo fue necesario.
- Cuántos recursos participaron.
- Qué dificultades aparecieron.
- Qué resultados se obtuvieron.

La lógica es sencilla:

**Proyecto anterior → Datos → Análisis → Estimación del proyecto nuevo**

Mientras más confiables sean los datos utilizados, mayor fundamento tendrá la estimación.

Esto no significa que el futuro será exactamente igual al pasado. Significa que los datos anteriores pueden servir como referencia para construir una previsión.

## 4. Parámetros y algoritmos

Después de obtener datos históricos se pueden identificar **parámetros cuantificables**.

Un parámetro puede ser, por ejemplo:

- Número de líneas de código.
- Tamaño del producto.
- Cantidad de funcionalidades.
- Número de elementos a desarrollar.
- Tamaño de una actividad.

Después se aplica un método o algoritmo que permite obtener una estimación.

La secuencia presentada puede resumirse:

**Datos históricos → Parámetro → Algoritmo → Estimación**

Esto permite pasar de una opinión general a una estimación basada en información.

## 5. Técnicas de estimación

El video presenta varias técnicas.

### 1. Estimación análoga

Compara el trabajo actual con un proyecto o actividad similar realizada anteriormente.

**Ejemplo:**

Si una funcionalidad similar tomó 5 días en un proyecto anterior, puede utilizarse como referencia para una nueva estimación, haciendo los ajustes necesarios.

### 2. Estimación paramétrica

Utiliza una relación cuantificable entre una medida y el resultado esperado.

**Ejemplo:**

Si históricamente se desarrollan 10 funcionalidades por semana, esa relación puede utilizarse como referencia para estimar un trabajo similar.

### 3. Estimación multipunto / PERT

Utiliza diferentes escenarios de estimación para considerar incertidumbre.

Normalmente se consideran:

- Optimista.
- Más probable.
- Pesimista.

Una fórmula habitual de PERT es:

**Estimación = (O + 4M + P) / 6**

donde:

- O = optimista.
- M = más probable.
- P = pesimista.

### 4. Estimación ascendente

Comienza con elementos pequeños y posteriormente los suma para obtener una estimación general.

**Parte pequeña → Subtotal → Total del proyecto**

## 6. Enfoque predictivo y adaptativo

El video muestra que la técnica de estimación puede cambiar según el enfoque de gestión.

### Predictivo

En un enfoque predictivo se pueden utilizar modelos más estructurados y basados en medidas del producto.

El video presenta como ejemplo **COCOMO**, que utiliza características relacionadas con el software, como las líneas de código, para realizar estimaciones.

### Adaptativo

En enfoques ágiles se pueden utilizar unidades como los **puntos de historia**.

Los puntos de historia representan una estimación relativa del tamaño o esfuerzo de una historia de usuario.

Por ejemplo:

**Historia A = 5 puntos**

**Historia B = 13 puntos**

No significa que 13 puntos sean exactamente 13 horas. Son una medida relativa utilizada por el equipo.

### Idea principal

**Predictivo → modelos y medidas más definidos**

**Adaptativo → estimación relativa y aprendizaje mediante iteraciones**

La técnica debe ser coherente con la naturaleza del trabajo y con la información disponible.

## 8. ¿Por qué una estimación no es una certeza?

Una estimación es una **previsión**, no una garantía.

El resultado puede cambiar por:

- Incertidumbre.
- Cambios en los requisitos.
- Disponibilidad de recursos.
- Complejidad real del trabajo.
- Problemas técnicos.
- Dependencias.
- Información incompleta.

Por eso una buena práctica consiste en revisar las estimaciones a medida que aparece nueva información.

La estimación puede mejorar mediante un ciclo:

**Estimar → Ejecutar → Medir → Aprender → Ajustar**

## 9. Análisis del video

La principal enseñanza del video es que estimar no significa simplemente decir cuánto creemos que tardará algo.

Una estimación profesional debe intentar responder:

1. ¿Qué información tenemos?
2. ¿Existe un trabajo anterior comparable?
3. ¿Qué parámetro podemos medir?
4. ¿Qué técnica es adecuada?
5. ¿Qué incertidumbre existe?
6. ¿Cómo justificamos el resultado?

Esto permite que la estimación sea entendible y pueda ser revisada.

También es importante diferenciar **esfuerzo** y **duración**, porque confundirlos puede generar errores en el cronograma.

## 12. Ejemplo sencillo de PERT

Supongamos una actividad de software con estas estimaciones:

- Optimista: 4 días.
- Más probable: 6 días.
- Pesimista: 10 días.

La fórmula es:

**PERT = (O + 4M + P) / 6**

Sustituyendo:

**PERT = (4 + 4(6) + 10) / 6**

**PERT = 38 / 6**

**PERT ≈ 6,33 días**

Este cálculo sirve como ejemplo de cómo considerar diferentes escenarios en una estimación.

El resultado es una estimación, no una duración garantizada.

## 13. Relación con Ingeniería de Proyectos

Las estimaciones son fundamentales porque afectan directamente:

- El cronograma.
- Los recursos.
- Los costos.
- Las fechas de entrega.
- La planificación del proyecto.

Una estimación demasiado optimista puede generar expectativas difíciles de cumplir.

Una estimación basada en datos y revisada durante el proyecto permite tomar decisiones con mejor información.

Por eso el gerente de proyectos debe comprender no solamente el resultado de una estimación, sino también **cómo fue obtenida y qué supuestos existen detrás de ella**.

## 14. Conclusión

El video presenta la estimación como una herramienta para construir una previsión razonada del futuro del proyecto.

La idea central puede resumirse así:

**Datos históricos → Parámetros → Técnica → Estimación → Seguimiento → Ajuste**

También deja una enseñanza importante:

> Esfuerzo y duración son conceptos diferentes y deben analizarse por separado.

Finalmente, el método de estimación debe adaptarse al tipo de proyecto. Un proyecto predictivo puede utilizar modelos paramétricos o técnicas tradicionales, mientras que un equipo adaptativo puede utilizar estimaciones relativas como puntos de historia.

Esto prepara el camino para estudiar cómo identificar las actividades que realmente determinan la duración total del proyecto.
