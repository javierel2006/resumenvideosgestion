# Dominio Financiero - Javier Moreno

## Análisis y resumen del video 9

**Asignatura:** Ingeniería de Proyectos

### Propósito
Comprender la importancia de la gestión financiera del proyecto y conocer conceptos básicos para analizar inversión, rentabilidad y recuperación de los costos.

## 1. Resumen del video

El video presenta el **Dominio Financiero** y explica que la gestión de un proyecto también debe considerar su dimensión económica.

El tema central es analizar si los recursos invertidos en un proyecto se relacionan con los beneficios y resultados que se esperan obtener.

Entre los conceptos destacados se encuentra el **ROI (Return on Investment)** o retorno de la inversión y el **periodo de recuperación**, que permite conocer cuánto tiempo se necesita para recuperar el costo inicial.

El video utiliza ejemplos relacionados con proyectos tecnológicos, servidores y servicios en la nube, mostrando que las decisiones financieras deben acompañar las decisiones técnicas.

Una idea general es:

**Inversión → Costos → Beneficios → Recuperación → Rentabilidad → Valor**

## 2. ¿Qué es el dominio financiero?

El dominio financiero se ocupa de la dimensión económica del proyecto.

No basta con saber:

**¿Qué vamos a construir?**

También es necesario conocer:

**¿Cuánto cuesta?**

**¿Qué inversión necesita?**

**¿Qué beneficios esperamos?**

**¿Cuándo podemos recuperar la inversión?**

En un proyecto de software pueden existir costos relacionados con:

- Personal.
- Equipos.
- Servidores.
- Servicios en la nube.
- Licencias.
- Herramientas.
- Implementación.
- Mantenimiento inicial.

La gestión financiera permite organizar esta información y utilizarla para apoyar las decisiones del proyecto.

## 3. Inversión y costos

La inversión inicial representa los recursos económicos necesarios para poner en marcha una iniciativa.

Por ejemplo, un proyecto de software podría necesitar inicialmente:

- Compra de equipos.
- Contratación del equipo.
- Configuración de infraestructura.
- Licencias.
- Desarrollo inicial.
- Servicios tecnológicos.

Después pueden aparecer costos durante la operación o ejecución.

Por eso es útil diferenciar:

**Costo inicial**

de

**Costos posteriores**

y relacionarlos con los beneficios esperados.

## 4. ROI — Retorno de la inversión

El video presenta el concepto de **ROI**, utilizado para medir la rentabilidad de una inversión.

Una fórmula básica es:

**ROI = (Beneficio neto / Inversión inicial) × 100**

Donde:

**Beneficio neto = Beneficios obtenidos - Inversión**

### Ejemplo

Supongamos:

- Inversión inicial = $10.000.000
- Beneficio obtenido = $13.000.000

Entonces:

**Beneficio neto = $13.000.000 - $10.000.000**

**Beneficio neto = $3.000.000**

Y:

**ROI = ($3.000.000 / $10.000.000) × 100**

**ROI = 30 %**

Este ejemplo muestra cómo relacionar el beneficio neto con la inversión realizada.

## 5. Periodo de recuperación

El video también presenta el **periodo de recuperación**, conocido como Payback.

Su objetivo es conocer cuánto tiempo se necesita para recuperar el costo inicial de la inversión.

Una forma sencilla de calcularlo, cuando los flujos son constantes, es:

**Payback = Inversión inicial / Flujo de efectivo periódico**

### Ejemplo

Si la inversión inicial es:

**$14.000.000**

y el proyecto genera:

**$1.000.000 mensuales**

entonces:

**Payback = $14.000.000 / $1.000.000**

**Payback = 14 meses**

El video presenta precisamente un ejemplo visual de **14 meses** para la recuperación de la inversión.

## 6. ¿Por qué importa el tiempo de recuperación?

El periodo de recuperación permite observar cuándo la inversión inicial ha sido compensada por los flujos obtenidos.

Puede representarse así:

**Mes 1 → inversión y primeros resultados**

↓

**Mes 6 → avance en la recuperación**

↓

**Mes 14 → recuperación de la inversión inicial**

A partir de este punto, el análisis puede continuar observando los beneficios posteriores.

Es importante recordar que el Payback es una medida de recuperación y no representa por sí solo todo el análisis económico de un proyecto.

## 7. Tecnología y decisiones financieras

El video utiliza elementos tecnológicos como servidores y servicios en la nube para mostrar que una decisión técnica también puede tener consecuencias económicas.

Por ejemplo, una organización puede comparar:

**Infraestructura propia**

frente a

**Servicios en la nube**

Cada alternativa puede implicar diferencias en:

- Inversión inicial.
- Costos periódicos.
- Capacidad.
- Mantenimiento.
- Escalabilidad.
- Tiempo de implementación.

Por eso las decisiones técnicas deben analizarse junto con sus consecuencias financieras.

## 8. Rentabilidad y recuperación no son lo mismo

Es importante distinguir dos preguntas:

### Rentabilidad

**¿Qué beneficio obtenemos en relación con la inversión?**

Aquí puede utilizarse el ROI como una medida básica.

### Recuperación

**¿Cuánto tiempo necesitamos para recuperar la inversión inicial?**

Aquí se utiliza el Payback.

Un proyecto puede recuperar rápidamente una inversión y, aun así, requerir un análisis más amplio para conocer su rentabilidad total.

Por eso ambos indicadores responden preguntas diferentes.

## 9. Análisis del video

La principal enseñanza del video es que la gestión financiera debe acompañar todo el proyecto.

Antes de iniciar una inversión es necesario comprender:

**Qué cuesta → Qué se espera obtener → Cuándo se recupera → Qué rentabilidad puede generar**

Esto conecta directamente con el concepto de **valor** estudiado en los videos anteriores.

Un proyecto no debería analizarse solamente por sus entregables técnicos. También es importante comprender sus consecuencias económicas y su contribución a los objetivos de la organización.

## 11. Conceptos para recordar

| Concepto | Explicación sencilla |
|---|---|
| Dominio financiero | Gestión de la dimensión económica del proyecto |
| Inversión inicial | Recursos económicos necesarios para iniciar |
| Costo | Recurso económico utilizado para realizar el proyecto |
| Beneficio | Resultado económico o de negocio que se espera obtener |
| ROI | Indicador que relaciona el beneficio neto con la inversión |
| Payback | Tiempo necesario para recuperar la inversión inicial |
| Rentabilidad | Capacidad de generar beneficios respecto a la inversión |
| Flujo de efectivo | Entradas y salidas de dinero durante un periodo |

## 12. Ejemplo completo

Supongamos un proyecto de software:

**Inversión inicial:** $20.000.000

**Beneficio total considerado:** $26.000.000

### Paso 1 — Beneficio neto

$26.000.000 - $20.000.000 = $6.000.000

### Paso 2 — ROI

ROI = ($6.000.000 / $20.000.000) × 100

ROI = **30 %**

### Paso 3 — Recuperación

Si el proyecto genera un flujo promedio de $2.000.000 mensuales:

Payback = $20.000.000 / $2.000.000

Payback = **10 meses**

Este ejercicio permite observar que ROI y Payback responden preguntas diferentes:

**ROI → cuánto representa el beneficio respecto a la inversión.**

**Payback → cuánto tiempo toma recuperar la inversión.**

## 13. Importancia para Ingeniería de Proyectos

El dominio financiero permite que el gerente de proyectos tenga una visión más completa.

Al planificar un proyecto de software se deben considerar:

- Costos de personal.
- Infraestructura.
- Licencias.
- Servicios tecnológicos.
- Costos de implementación.
- Beneficios esperados.
- Flujo de efectivo.
- Tiempo de recuperación.

Esto ayuda a relacionar las decisiones técnicas con las necesidades económicas de la organización.

## 14. Conclusión

El video muestra que un proyecto también debe analizarse desde el punto de vista financiero.

La idea puede resumirse:

**Inversión → Costos → Beneficios → ROI → Recuperación → Valor**

El **ROI** ayuda a analizar la relación entre beneficio neto e inversión.

El **Payback** ayuda a conocer el tiempo necesario para recuperar la inversión inicial.

Ambos conceptos son útiles, pero no deben confundirse.

La enseñanza principal es que las decisiones del proyecto deben considerar simultáneamente el trabajo técnico, los recursos, el tiempo y la dimensión económica.
