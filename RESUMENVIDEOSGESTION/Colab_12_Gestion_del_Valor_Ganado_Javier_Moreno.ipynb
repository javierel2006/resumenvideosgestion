## 1. Resumen del video

La Gestión del Valor Ganado permite medir cómo avanza un proyecto comparando tres elementos:

- Lo que estaba planeado.
- El valor del trabajo que realmente se ha completado.
- El costo que realmente se ha gastado.

La herramienta ayuda a responder preguntas sencillas:

**¿Cuánto trabajo debía haber realizado?**

**¿Cuánto trabajo he realizado realmente?**

**¿Cuánto dinero he gastado para hacerlo?**

Con estas comparaciones se pueden detectar retrasos y sobrecostos antes de finalizar el proyecto.

## 2. Los tres valores principales

### PV - Valor Planificado

Es el valor del trabajo que estaba previsto realizar hasta una fecha determinada.

También puede encontrarse como **Planned Value**.

### EV - Valor Ganado

Es el valor presupuestado del trabajo que realmente se ha completado.

También se conoce como **Earned Value**.

### AC - Costo Real

Es el dinero que realmente se ha gastado para realizar el trabajo.

También se conoce como **Actual Cost**.

La relación básica es:

**PV = trabajo planificado**

**EV = trabajo realizado expresado en valor**

**AC = costo realmente gastado**

## 3. Ejemplo sencillo

Supongamos un proyecto con un presupuesto total de:

**$100.000**

Después de cierto tiempo:

- Se esperaba haber realizado trabajo por **$60.000**.
- El trabajo realmente completado tiene un valor de **$50.000**.
- El costo real gastado fue de **$55.000**.

Entonces:

**PV = $60.000**

**EV = $50.000**

**AC = $55.000**

Estos tres datos permiten analizar el estado del proyecto.

## 4. Variación del cronograma

La variación del cronograma puede calcularse mediante:

**SV = EV - PV**

Con el ejemplo:

**SV = $50.000 - $60.000**

**SV = -$10.000**

Un valor negativo indica que el valor del trabajo realizado está por debajo de lo que se había planificado para ese momento.

Por tanto, existe una desviación relacionada con el cronograma.

## 5. Variación del costo

La variación del costo se calcula:

**CV = EV - AC**

Con el ejemplo:

**CV = $50.000 - $55.000**

**CV = -$5.000**

El resultado negativo indica que se ha gastado más dinero del que corresponde al valor del trabajo realizado.

Esto muestra una desviación de costo.

## 6. Índice de desempeño del cronograma

El **SPI** permite analizar el desempeño respecto al cronograma.

Fórmula:

**SPI = EV / PV**

Con el ejemplo:

**SPI = 50.000 / 60.000**

**SPI = 0,83**

Una interpretación sencilla:

- **SPI = 1:** el desempeño coincide con lo planificado.
- **SPI > 1:** el valor realizado está por encima de lo planificado.
- **SPI < 1:** el valor realizado está por debajo de lo planificado.

En este caso, **0,83** muestra un desempeño inferior al planificado.

## 7. Índice de desempeño del costo

El **CPI** permite analizar qué tan eficientemente se está utilizando el dinero.

Fórmula:

**CPI = EV / AC**

Con el ejemplo:

**CPI = 50.000 / 55.000**

**CPI = 0,91**

Interpretación sencilla:

- **CPI = 1:** desempeño económico según lo planificado.
- **CPI > 1:** el valor obtenido es mayor que el costo utilizado.
- **CPI < 1:** el costo utilizado es mayor en relación con el valor obtenido.

En este ejemplo, **0,91** indica una eficiencia de costo inferior a la referencia.

## 8. Tabla de análisis

Los indicadores no sustituyen el análisis del gerente. Sirven como señales para investigar qué está sucediendo.

## 10. Curva S

La Gestión del Valor Ganado también puede representarse mediante una **Curva S**.

En ella se pueden comparar:

- Valor Planificado (PV).
- Valor Ganado (EV).
- Costo Real (AC).

El gráfico permite observar visualmente cómo evoluciona el proyecto a lo largo del tiempo.

Cuando las curvas se separan, puede existir una diferencia que debe ser analizada.

```python
# Ejemplo sencillo de Valor Ganado
import matplotlib.pyplot as plt

periodos = ["Inicio", "Mes 1", "Mes 2", "Mes 3", "Mes 4"]
pv = [0, 20, 45, 70, 100]
ev = [0, 15, 38, 58, 82]
ac = [0, 22, 48, 68, 95]

plt.figure(figsize=(10, 5))
plt.plot(periodos, pv, marker="o", label="PV - Planificado")
plt.plot(periodos, ev, marker="o", label="EV - Ganado")
plt.plot(periodos, ac, marker="o", label="AC - Costo Real")

plt.title("Ejemplo de Curva S - Valor Ganado")
plt.xlabel("Periodo")
plt.ylabel("Valor")
plt.grid(True)
plt.legend()
plt.show()
```

## 11. ¿Para qué sirve el Valor Ganado?

Permite:

- Detectar retrasos.
- Detectar sobrecostos.
- Medir el desempeño.
- Comparar resultados con la línea base.
- Obtener indicadores.
- Apoyar la toma de decisiones.
- Anticipar problemas del proyecto.

Su principal ventaja es que reúne información de diferentes áreas en una misma evaluación.

## 12. Ejemplo aplicado a un proyecto de software

Supongamos el desarrollo de una aplicación.

Presupuesto total:

**$50.000.000**

Después de varias semanas:

**PV = $30.000.000**

**EV = $25.000.000**

**AC = $28.000.000**

Entonces:

**SV = EV - PV = $25.000.000 - $30.000.000 = -$5.000.000**

**CV = EV - AC = $25.000.000 - $28.000.000 = -$3.000.000**

**SPI = 25 / 30 = 0,83**

**CPI = 25 / 28 = 0,89**

Los resultados indican que el proyecto debe ser revisado porque el trabajo realizado está por debajo del plan y el costo utilizado es superior al valor obtenido.

## 13. Análisis del resultado

Los indicadores permiten identificar una situación, pero después es necesario buscar la causa.

Por ejemplo, un SPI menor que 1 podría estar relacionado con:

- Retrasos en requisitos.
- Falta de recursos.
- Problemas técnicos.
- Dependencias entre actividades.
- Cambios de alcance.

Un CPI menor que 1 podría relacionarse con:

- Mayor cantidad de horas trabajadas.
- Recursos más costosos.
- Retrabajo.
- Problemas de calidad.
- Estimaciones iniciales insuficientes.

Por eso el Valor Ganado es una herramienta de análisis y no solamente una operación matemática.
