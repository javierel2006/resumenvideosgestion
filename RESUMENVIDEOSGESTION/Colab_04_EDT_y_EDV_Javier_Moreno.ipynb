# EDT y EDV - Javier Moreno

## Análisis y resumen del video 4

**Asignatura:** Ingeniería de Proyectos

### Propósito
Comprender cómo la **EDT** organiza y descompone el trabajo del proyecto y cómo la **EDV** relaciona ese trabajo con el valor de negocio esperado.

## 1. Resumen del video

El video explica dos estructuras importantes para organizar un proyecto:

- **EDT:** Estructura de Descomposición del Trabajo.
- **EDV:** Estructura de Descomposición del Valor.

La **EDT** permite tomar el alcance del proyecto y dividirlo de forma jerárquica hasta llegar a elementos que puedan ser gestionados, estimados y controlados.

La **EDV** complementa esta visión porque conecta el alcance del proyecto con el **valor de negocio esperado**.

En términos sencillos:

**EDT = ¿Qué trabajo debemos organizar y entregar?**

**EDV = ¿Cómo se relaciona ese trabajo con el valor que esperamos obtener?**

## 2. ¿Qué es la EDT?

La EDT es una **descomposición jerárquica del alcance**.

Su finalidad es dividir el proyecto en partes cada vez más pequeñas y comprensibles hasta llegar a elementos que puedan ser gestionados.

El video utiliza un ejemplo de construcción para mostrar la idea:

**Proyecto**
↓
**Sector**
↓
**Distrito**
↓
**Edificio**

La misma lógica puede utilizarse en un proyecto de software.

Por ejemplo:

**Sistema de ventas**
↓
**Módulo de usuarios**
↓
**Autenticación**
↓
**Inicio de sesión**

## 3. Regla del 100 %

Un concepto importante presentado en el video es la **regla del 100 %**.

La idea es que la EDT debe representar el **100 % del trabajo necesario para completar el alcance definido**.

Por lo tanto, al dividir un elemento en componentes, esos componentes deben cubrir todo lo necesario de ese nivel.

### Ejemplo sencillo

Si el proyecto consiste en desarrollar una aplicación web, la EDT podría incluir:

- Análisis de requisitos.
- Diseño.
- Desarrollo.
- Pruebas.
- Implementación.
- Documentación.

La suma de estos componentes debe representar el trabajo incluido en el alcance del proyecto.

La regla ayuda a evitar dos problemas:

**Omitir trabajo** o **agregar trabajo que no pertenece al alcance**.

## 4. Del proyecto al paquete de trabajo

La EDT permite pasar progresivamente de una visión general a elementos concretos.

Una secuencia puede ser:

**Producto**
→ **Planificación**
→ **Cuenta de control**
→ **Paquete de trabajo**

### Paquete de trabajo

Es un nivel suficientemente detallado para poder estimar y gestionar aspectos como:

- Costos.
- Recursos.
- Tiempo.
- Trabajo necesario.

Por ejemplo, dentro de un proyecto de software:

**Sistema**
→ **Módulo de usuarios**
→ **Autenticación**
→ **Paquete de trabajo: implementar inicio de sesión**

Esto facilita posteriormente la estimación y la elaboración del cronograma.

## 5. ¿Qué es la EDV?

La **EDV** es una estructura jerárquica que conecta el **alcance del proyecto con el valor de negocio esperado**.

Su función es ayudar a responder:

**¿Por qué estamos realizando este trabajo?**

Mientras la EDT se concentra principalmente en organizar el trabajo y los entregables, la EDV ayuda a relacionar esos elementos con los resultados y beneficios que se esperan.

### Relación sencilla

**Trabajo**
↓
**Entregable**
↓
**Resultado**
↓
**Beneficio**
↓
**Valor de negocio**

## 6. Diferencia entre EDT y EDV

| Estructura | Pregunta principal | Enfoque |
|---|---|---|
| EDT | ¿Qué trabajo debemos realizar? | Trabajo y entregables |
| EDV | ¿Qué valor esperamos generar? | Resultados, beneficios y valor |

No son estructuras que compitan entre sí.

La EDT ayuda a organizar el proyecto y la EDV permite mantener visible la relación entre lo que se hace y el valor que se espera conseguir.

## 7. Ejemplo aplicado a un proyecto de software

Supongamos un proyecto para desarrollar un sistema de ventas.

### EDT

**Proyecto: Sistema de ventas**

- 1. Análisis
- 2. Diseño
- 3. Desarrollo
  - 3.1 Usuarios
  - 3.2 Productos
  - 3.3 Ventas
  - 3.4 Reportes
- 4. Pruebas
- 5. Implementación
- 6. Capacitación

Esta estructura muestra **qué trabajo y entregables** forman parte del proyecto.

### EDV

Ese mismo proyecto puede relacionarse con:

**Sistema de ventas**
→ Automatización del proceso
→ Menos trabajo manual
→ Información disponible más rápidamente
→ Mejor apoyo para la toma de decisiones
→ Valor para la organización

Aquí observamos **por qué ese trabajo tiene importancia para el negocio**.

## 8. Análisis del video

La principal enseñanza es que un proyecto necesita una estructura clara.

Cuando un proyecto es demasiado grande para gestionarlo como un solo bloque, se divide en componentes.

La EDT permite responder:

> ¿Cómo podemos dividir el alcance para organizar y controlar el trabajo?

La EDV agrega otra pregunta:

> ¿Cómo se relaciona ese trabajo con los resultados y el valor que esperamos obtener?

Esto ayuda a evitar que el equipo se concentre únicamente en completar tareas sin recordar el propósito del proyecto.

## 10. Importancia para Ingeniería de Proyectos

La EDT es importante porque permite:

- Organizar el trabajo.
- Dividir proyectos complejos.
- Asignar responsabilidades.
- Estimar costos.
- Estimar recursos.
- Preparar el cronograma.
- Facilitar el seguimiento.

La EDV ayuda a mantener una conexión entre las actividades del proyecto y el propósito empresarial.

En conjunto, permiten observar el proyecto desde dos perspectivas:

**Trabajo que se debe realizar + valor que se espera generar**.

## 11. Conceptos para recordar

| Concepto | Explicación sencilla |
|---|---|
| EDT | Descompone jerárquicamente el alcance y el trabajo |
| Regla del 100 % | La EDT debe cubrir todo el trabajo incluido en el alcance |
| Paquete de trabajo | Nivel detallado que permite estimar y gestionar el trabajo |
| Cuenta de control | Punto donde se puede integrar y controlar el desempeño |
| EDV | Relaciona el alcance con el valor de negocio esperado |
| Entregable | Resultado que debe producir el proyecto |
| Beneficio | Resultado útil obtenido |
| Valor | Importancia o utilidad generada por los resultados del proyecto |

## 12. Conclusión

La EDT permite convertir un proyecto grande en partes organizadas y manejables.

La EDV permite no perder de vista el propósito de ese trabajo, conectándolo con los resultados, beneficios y valor esperado.

La idea principal puede resumirse así:

**EDT → organiza el trabajo**

**EDV → conecta el trabajo con el valor**

Ambas estructuras ayudan a que la planificación posterior sea más clara y sirven como base para trabajar con el cronograma, los recursos y las estimaciones.

## 13. Actividad de análisis

Para un proyecto de software que conozcas, identifica:

### EDT
1. Producto principal.
2. Tres entregables principales.
3. Dos elementos que puedan formar parte de cada entregable.
4. Un posible paquete de trabajo.

### EDV
1. ¿Qué resultado se espera?
2. ¿Qué beneficio puede producir?
3. ¿Qué valor tendría para la organización?

La finalidad es practicar la diferencia entre **organizar el trabajo** y **relacionar ese trabajo con el valor esperado**.
