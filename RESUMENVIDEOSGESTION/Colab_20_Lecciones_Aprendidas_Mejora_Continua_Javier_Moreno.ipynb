### Propósito
Comprender cómo aprovechar la experiencia obtenida durante un proyecto para mejorar el trabajo actual y los futuros proyectos.

## 1. Resumen del tema

Durante un proyecto ocurren situaciones que dejan conocimientos útiles.

Algunas cosas funcionan muy bien y otras necesitan mejorar.

Las **lecciones aprendidas** permiten registrar esas experiencias para que no se pierdan.

La **mejora continua** utiliza ese conocimiento para cambiar y mejorar la forma de trabajar.

La idea principal es:

**Aprender de la experiencia → aplicar lo aprendido → mejorar los resultados.**

## 2. ¿Qué es una lección aprendida?

Es un conocimiento obtenido a partir de una experiencia del proyecto.

Puede surgir de:

- Un problema.
- Un error.
- Una buena práctica.
- Una decisión.
- Un cambio.
- Un éxito.
- Una situación inesperada.

Ejemplo:

> El equipo descubrió que revisar los requisitos con el cliente antes de programar redujo los cambios posteriores.

Esta experiencia puede convertirse en una recomendación para futuros proyectos.

## 3. Las lecciones no son solamente problemas

Una lección aprendida también puede registrar algo que funcionó correctamente.

### Ejemplo positivo

> Las reuniones cortas diarias ayudaron a detectar bloqueos rápidamente.

### Ejemplo de mejora

> No definir claramente los criterios de aceptación generó retrabajo.

Ambas experiencias son útiles porque ayudan a repetir buenas prácticas y evitar errores.

## 4. ¿Cuándo se deben registrar?

Las lecciones aprendidas pueden registrarse durante todo el proyecto.

No es recomendable esperar únicamente al cierre.

Se pueden identificar después de:

- Una reunión importante.
- Un problema.
- Una entrega.
- Una revisión.
- Una prueba.
- Un cambio.
- Una fase del proyecto.
- Un Sprint.

Así el conocimiento permanece disponible mientras todavía puede ser utilizado.

## 5. Registro de lecciones aprendidas

Un registro sencillo puede contener:

| Fecha | Situación | Lección | Acción futura |
|---|---|---|---|
| Semana 1 | Requisitos incompletos | Validar requisitos antes de desarrollar | Crear revisión inicial |
| Semana 3 | Pruebas tardías | Probar desde etapas tempranas | Incorporar pruebas continuas |
| Semana 5 | Buena comunicación | Reuniones breves facilitaron el avance | Mantener reuniones |

## 6. Mejora continua

La mejora continua consiste en revisar cómo se está trabajando y buscar oportunidades para hacerlo mejor.

Una secuencia sencilla es:

**Planificar → Ejecutar → Revisar → Mejorar**

Después de aplicar una mejora se vuelve a revisar el resultado.

De esta manera el proceso puede evolucionar progresivamente.

## 7. Ciclo de mejora

Un ejemplo muy conocido es el ciclo:

### Planificar
Definir qué se quiere mejorar.

### Hacer
Aplicar el cambio.

### Verificar
Comprobar los resultados.

### Actuar
Mantener la mejora o realizar nuevos ajustes.

Este ciclo puede repetirse varias veces durante el proyecto.

## 8. Ejemplo sencillo

Un equipo tarda demasiado en corregir errores.

### Situación
Los defectos se detectan al final de cada entrega.

### Mejora propuesta
Realizar pruebas durante el desarrollo.

### Aplicación
Se incorporan pruebas más frecuentes.

### Resultado
Los errores se detectan antes.

### Nueva lección
Las pruebas tempranas reducen el retrabajo.

La experiencia puede utilizarse nuevamente en futuros proyectos.

```python
# Ejemplo sencillo de registro de lecciones aprendidas

lecciones = [
    ("Revisar requisitos temprano", "Reduce cambios posteriores"),
    ("Probar continuamente", "Permite detectar errores antes"),
    ("Comunicación frecuente", "Ayuda a identificar bloqueos")
]

for numero, (leccion, beneficio) in enumerate(lecciones, start=1):
    print(f"{numero}. {leccion} -> {beneficio}")
```

## 9. Causa y solución

Una buena lección aprendida no debería limitarse a decir:

> "Hubo retrasos."

Es mejor identificar:

**Qué ocurrió → Por qué ocurrió → Qué se hizo → Qué se debe hacer en el futuro.**

Ejemplo:

**Qué ocurrió:** una funcionalidad se retrasó.

**Causa:** el requisito no estaba suficientemente definido.

**Acción:** realizar una revisión con el cliente.

**Lección:** validar criterios de aceptación antes de comenzar el desarrollo.

## 10. Compartir el conocimiento

Las lecciones aprendidas deben quedar disponibles para las personas que puedan utilizarlas.

Pueden almacenarse en:

- Documentos.
- Repositorios.
- Wikis.
- Sistemas de gestión.
- Informes de proyecto.
- Bases de conocimiento.

El objetivo es evitar que el conocimiento desaparezca cuando el proyecto termina o cuando cambian los integrantes del equipo.

## 11. Mejora continua en proyectos de software

En software puede aplicarse para mejorar:

- Desarrollo.
- Pruebas.
- Requisitos.
- Comunicación.
- Estimaciones.
- Despliegues.
- Seguridad.
- Documentación.
- Gestión de incidencias.

Por ejemplo, si cada Sprint presenta el mismo tipo de error, el equipo puede investigar la causa y modificar su proceso.

## 12. Retrospectivas

En equipos ágiles, una retrospectiva permite revisar cómo se desarrolló el trabajo.

Preguntas sencillas:

- ¿Qué salió bien?
- ¿Qué salió mal?
- ¿Qué debemos mantener?
- ¿Qué debemos cambiar?
- ¿Qué acción concreta realizaremos?

La retrospectiva convierte la experiencia del equipo en oportunidades de mejora.

## 14. Ejemplo aplicado

En un proyecto de aplicación web se detecta que varias funcionalidades tuvieron retrasos porque los criterios de aceptación no estaban claros.

Después de analizar la situación, el equipo decide:

1. Revisar los criterios con el cliente.
2. Documentarlos antes de iniciar el desarrollo.
3. Validarlos durante cada Sprint.
4. Registrar la experiencia como lección aprendida.

La mejora se convierte así en una nueva práctica de trabajo.

Un proyecto no solamente produce un producto o servicio.

Si ese conocimiento se registra y se utiliza, la organización puede mejorar sus procesos y aumentar su capacidad para realizar futuros proyectos.

La secuencia puede resumirse:

**Experiencia → Análisis → Lección → Acción → Mejora → Nueva experiencia**

Por eso, aprender durante el proyecto es tan importante como completar las actividades planificadas.
