### Propósito
Comprender cómo planificar, seleccionar y controlar las compras y servicios externos necesarios para desarrollar un proyecto.

## 1. Resumen del tema

No todos los recursos de un proyecto tienen que ser proporcionados directamente por el equipo.

En muchas ocasiones es necesario adquirir:

- Software.
- Equipos.
- Servicios en la nube.
- Licencias.
- Consultorías.
- Servicios especializados.
- Materiales.
- Servicios de soporte.

La gestión de adquisiciones permite organizar estas compras para obtener lo necesario, en las condiciones acordadas y en el momento requerido.

## 2. ¿Qué es una adquisición?

Una adquisición es el proceso mediante el cual el proyecto obtiene productos, servicios o recursos de una fuente externa.

Por ejemplo:

> Una empresa necesita contratar un servicio de alojamiento en la nube para poner en funcionamiento una aplicación.

En este caso existe una necesidad del proyecto que será cubierta mediante un proveedor externo.

## 3. Comprar o hacer

Una decisión importante consiste en determinar si algo debe:

**Hacerse internamente**

o

**Adquirirse externamente.**

Para tomar la decisión se pueden considerar:

- Costo.
- Tiempo.
- Capacidad del equipo.
- Experiencia disponible.
- Calidad esperada.
- Riesgos.
- Seguridad.
- Dependencia del proveedor.

No siempre la opción más barata es la más conveniente para el proyecto.

## 4. Proceso sencillo de adquisición

Un proceso básico puede ser:

1. Identificar la necesidad.
2. Definir qué se necesita.
3. Buscar posibles proveedores.
4. Solicitar propuestas o cotizaciones.
5. Comparar alternativas.
6. Seleccionar el proveedor.
7. Formalizar el acuerdo.
8. Recibir y verificar el producto o servicio.
9. Realizar seguimiento.
10. Cerrar la adquisición.

## 5. Requisitos de la adquisición

Antes de buscar un proveedor es necesario definir claramente qué se necesita.

Por ejemplo, para contratar un servicio en la nube:

- Capacidad.
- Disponibilidad.
- Seguridad.
- Soporte.
- Tiempo de respuesta.
- Precio.
- Condiciones de servicio.

Mientras más clara sea la necesidad, más fácil será comparar propuestas.

## 6. Proveedores

Un proveedor puede entregar un producto, servicio o solución necesaria para el proyecto.

Al evaluar un proveedor pueden considerarse:

- Experiencia.
- Precio.
- Calidad.
- Capacidad técnica.
- Tiempo de entrega.
- Soporte.
- Reputación.
- Condiciones contractuales.
- Riesgos asociados.

La selección debe basarse en los requisitos definidos para la adquisición.

## 7. Contratos

El contrato o acuerdo establece las condiciones de la relación con el proveedor.

Puede incluir:

- Qué se entregará.
- Precio.
- Fechas.
- Responsabilidades.
- Criterios de aceptación.
- Garantías.
- Forma de pago.
- Condiciones de cambio.
- Soporte.
- Terminación del acuerdo.

Un acuerdo claro reduce confusiones posteriores.

## 8. Ejemplo sencillo

Un proyecto de software necesita contratar una empresa para realizar pruebas de seguridad.

Se reciben tres propuestas:

| Proveedor | Precio | Tiempo | Experiencia |
|---|---:|---:|---|
| A | $8.000.000 | 20 días | Alta |
| B | $6.000.000 | 30 días | Media |
| C | $9.000.000 | 15 días | Alta |

La decisión no debería basarse solamente en el precio.

También deben revisarse el tiempo, la experiencia, el alcance del servicio y los demás requisitos del proyecto.

## 9. Seguimiento de adquisiciones

Después de contratar un proveedor, el trabajo continúa.

Se debe revisar:

- Si cumple las fechas.
- Si entrega lo acordado.
- Si cumple la calidad.
- Si respeta el presupuesto.
- Si existen cambios.
- Si aparecen riesgos.
- Si es necesario realizar acciones correctivas.

Contratar un proveedor no significa dejar de controlar la adquisición.

```python
# Comparación sencilla de proveedores

proveedores = [
    ("Proveedor A", 8_000_000, 20, 5),
    ("Proveedor B", 6_000_000, 30, 3),
    ("Proveedor C", 9_000_000, 15, 5)
]

print("Proveedor | Precio | Días | Experiencia")
print("-" * 42)

for nombre, precio, dias, experiencia in proveedores:
    print(f"{nombre} | ${precio:,.0f} | {dias} | {experiencia}/5")
```

## 10. Riesgos de las adquisiciones

Las adquisiciones pueden generar riesgos como:

- Retrasos del proveedor.
- Aumento de precios.
- Incumplimiento.
- Problemas de calidad.
- Dependencia excesiva.
- Problemas de seguridad.
- Cambios en las condiciones del servicio.
- Falta de soporte.

Por eso las adquisiciones también deben relacionarse con la gestión de riesgos.

## 11. Adquisiciones en proyectos de software

En software son frecuentes las adquisiciones de:

- Servicios cloud.
- Dominios.
- Certificados.
- Licencias.
- APIs.
- Herramientas de desarrollo.
- Servicios de diseño.
- Consultorías.
- Pruebas especializadas.
- Servicios de seguridad.

Antes de contratar es importante revisar cómo la adquisición afecta el presupuesto, cronograma, seguridad y continuidad del proyecto.

## 12. Relación con los temas anteriores

La gestión de adquisiciones se relaciona con:

**Alcance**
→ define qué producto o servicio se necesita.

**Cronograma**
→ determina cuándo debe estar disponible.

**Costos**
→ incorpora el valor de la adquisición.

**Riesgos**
→ analiza posibles problemas con proveedores.

**Calidad**
→ establece criterios para aceptar lo recibido.

**Recursos**
→ permite decidir qué será interno y qué será externo.

**Interesados**
→ considera a las personas involucradas en la compra y el proveedor.

## 13. Importancia de los criterios de aceptación

Antes de recibir un producto o servicio es conveniente definir cómo se comprobará que cumple lo acordado.

Ejemplo:

Para una aplicación desarrollada por un proveedor:

- Debe incluir las funcionalidades acordadas.
- Debe cumplir los requisitos de seguridad.
- Debe superar las pruebas definidas.
- Debe entregar la documentación.
- Debe cumplir la fecha acordada.

Esto facilita determinar si la entrega puede ser aceptada.

## 14. Conceptos para recordar

| Proveedor | Persona u organización que entrega el producto o servicio |

## 15. Importancia para Ingeniería de Proyectos

Una adquisición mal planificada puede afectar directamente:

**Costo + Tiempo + Calidad + Riesgo**

Por eso el gerente del proyecto debe analizar la necesidad antes de contratar y continuar realizando seguimiento después de la contratación.

La adquisición forma parte de la planificación general del proyecto.

## 16. Conclusión

La gestión de adquisiciones permite obtener productos y servicios externos de forma organizada.

El proceso puede resumirse:

**Necesidad → Requisitos → Proveedores → Evaluación → Contrato → Seguimiento → Aceptación → Cierre**

Una buena adquisición debe estar alineada con los objetivos, presupuesto, cronograma, calidad y riesgos del proyecto.
