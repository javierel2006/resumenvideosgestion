## 1. Resumen del tema

Los enfoques ágiles buscan desarrollar productos de manera progresiva, entregando resultados en períodos cortos y permitiendo recibir comentarios de los usuarios.

A diferencia de un proyecto completamente planificado desde el comienzo, un enfoque adaptativo permite ajustar el trabajo a medida que se obtiene nueva información.

Esto resulta especialmente útil en proyectos de software donde los requisitos pueden cambiar durante el desarrollo.

## 2. ¿Qué es Agile?

Agile es una forma de trabajar que prioriza:

- Entregas frecuentes.
- Colaboración.
- Comunicación.
- Retroalimentación.
- Adaptación al cambio.
- Valor para el usuario.

La idea no es eliminar la planificación, sino realizarla de forma progresiva y ajustarla cuando sea necesario.

## 3. Predictivo y adaptativo

### Enfoque predictivo

Se intenta definir desde el inicio gran parte del alcance, cronograma y costos.

Es útil cuando los requisitos son relativamente estables.

### Enfoque adaptativo

El trabajo se desarrolla en ciclos y el alcance puede ajustarse según los resultados y comentarios recibidos.

Es útil cuando existe incertidumbre o los requisitos pueden cambiar.

También pueden combinarse ambos enfoques en un proyecto híbrido.

## 4. Principios básicos

Un enfoque ágil busca:

**Entregar valor temprano.**

**Trabajar de manera colaborativa.**

**Obtener retroalimentación frecuente.**

**Responder a los cambios.**

**Mejorar continuamente.**

**Mantener comunicación constante con los interesados.**

## 5. Trabajo por iteraciones

Una iteración es un período corto en el que el equipo realiza un conjunto de actividades y produce un resultado.

Ejemplo:

**Iteración 1**
→ Inicio de sesión.

**Iteración 2**
→ Registro de usuarios.

**Iteración 3**
→ Perfil del usuario.

**Iteración 4**
→ Reportes.

En lugar de esperar hasta el final para mostrar todo el sistema, se van entregando partes funcionales.

## 6. Scrum como ejemplo

Scrum es uno de los marcos de trabajo ágiles más conocidos.

Algunos elementos básicos son:

- Product Backlog.
- Sprint.
- Sprint Backlog.
- Incremento.
- Revisión.
- Retrospectiva.

El trabajo se organiza en períodos llamados **Sprints**, durante los cuales el equipo busca producir un incremento de valor.

## 7. Roles de Scrum

De forma sencilla, Scrum contempla:

### Product Owner
Representa las necesidades del producto y ayuda a priorizar el trabajo.

### Scrum Master
Ayuda a que el marco de trabajo se aplique correctamente y facilita la eliminación de obstáculos.

### Developers
Construyen el incremento del producto.

La colaboración entre estos roles es fundamental.

## 8. Product Backlog

El Product Backlog contiene el trabajo que podría realizarse para mejorar el producto.

Puede incluir:

- Funcionalidades.
- Mejoras.
- Correcciones.
- Necesidades técnicas.
- Requisitos.

El contenido se prioriza de acuerdo con el valor, las necesidades y la información disponible.

## 9. Historias de usuario

Una historia de usuario expresa una necesidad desde la perspectiva del usuario.

Una estructura sencilla es:

**Como [tipo de usuario], quiero [necesidad], para [beneficio].**

Ejemplo:

> Como cliente, quiero consultar mis pedidos para conocer el estado de mis compras.

Esto ayuda a conectar las actividades de desarrollo con una necesidad real.

```python
# Ejemplo sencillo de Product Backlog

backlog = [
    ("Iniciar sesión", 1),
    ("Registrar usuario", 2),
    ("Consultar pedidos", 3),
    ("Generar reporte", 4),
    ("Editar perfil", 5)
]

print("Product Backlog")
print("-" * 30)

for tarea, prioridad in backlog:
    print(f"Prioridad {prioridad}: {tarea}")
```

## 10. Sprint

Un Sprint es un período de trabajo durante el cual el equipo desarrolla un incremento.

Durante el Sprint se:

1. Selecciona el trabajo.
2. Planifica.
3. Desarrolla.
4. Prueba.
5. Revisa el resultado.
6. Obtiene retroalimentación.
7. Identifican mejoras.

Después comienza un nuevo ciclo.

## 11. Retroalimentación

La retroalimentación es una parte importante de los enfoques adaptativos.

El equipo puede descubrir que:

- Una función no es necesaria.
- Una función necesita modificarse.
- El usuario necesita algo diferente.
- Existe un problema técnico.
- Una prioridad cambió.

Esta información puede utilizarse para ajustar el trabajo siguiente.

## 12. Priorización

No todo el trabajo tiene la misma importancia.

Para priorizar se pueden considerar:

- Valor para el usuario.
- Urgencia.
- Riesgo.
- Dependencias.
- Esfuerzo.
- Impacto en el negocio.

El objetivo es concentrar primero el trabajo que aporta mayor valor dentro del contexto del proyecto.

## 13. Estimación ágil

En proyectos ágiles se pueden utilizar técnicas como:

- Story Points.
- Tallas relativas.
- Planning Poker.
- Velocidad del equipo.

Los Story Points no representan directamente horas. Se utilizan para comparar el tamaño o complejidad relativa de las historias.

Ejemplo:

- Historia A = 2 puntos.
- Historia B = 5 puntos.
- Historia C = 8 puntos.

La historia C se considera relativamente mayor que las otras dos.

## 14. Mejora continua

Al finalizar una iteración, el equipo puede analizar:

- Qué funcionó.
- Qué no funcionó.
- Qué problemas aparecieron.
- Qué puede hacerse mejor.

Luego se seleccionan acciones concretas de mejora.

La idea es aprender durante el proyecto y no esperar hasta el cierre para descubrir todos los problemas.

## 15. Enfoque híbrido

No todos los proyectos deben ser completamente predictivos o completamente ágiles.

Un proyecto puede combinar ambos enfoques.

Ejemplo:

**Predictivo**
→ presupuesto general, contratos y fechas principales.

**Adaptativo**
→ desarrollo de funcionalidades mediante Sprints.

Este enfoque puede ser útil cuando algunas partes del proyecto son estables y otras tienen alta incertidumbre.

## 16. Relación con los temas anteriores

Los enfoques ágiles se relacionan con:

**Requisitos**
→ pueden evolucionar mediante retroalimentación.

**Cronograma**
→ se organiza mediante iteraciones o Sprints.

**Valor**
→ se busca entregar resultados útiles de forma frecuente.

**Riesgos**
→ se pueden descubrir problemas temprano.

**Calidad**
→ las pruebas se realizan continuamente.

**Interesados**
→ participan mediante comentarios y revisiones.

**Cambios**
→ se incorporan mediante priorización y planificación progresiva.

## 17. Ejemplo aplicado a una aplicación web

Un equipo debe desarrollar una aplicación de ventas.

En lugar de construir todo el sistema antes de mostrarlo:

**Sprint 1**
→ registro e inicio de sesión.

**Sprint 2**
→ catálogo de productos.

**Sprint 3**
→ carrito de compras.

**Sprint 4**
→ pagos.

Después de cada Sprint se revisa el resultado y se reciben comentarios.

Si los usuarios solicitan un cambio, el Product Backlog puede actualizarse y priorizarse para los siguientes ciclos.

## 19. Importancia para Ingeniería de Proyectos

Los proyectos de software suelen trabajar con cambios, incertidumbre y necesidades que evolucionan.

Los enfoques adaptativos permiten aprender durante el desarrollo y utilizar esa información para ajustar las prioridades.

La secuencia puede resumirse:

**Planificar → Desarrollar → Entregar → Revisar → Aprender → Adaptar**

## 20. Conclusión

Los enfoques ágiles y adaptativos permiten desarrollar productos de manera progresiva, recibir retroalimentación y responder a los cambios.

No significan trabajar sin planificación. Significan planificar de forma progresiva y utilizar la información obtenida para mejorar las siguientes decisiones.

En proyectos de software, esta forma de trabajo permite acercar continuamente el producto a las necesidades reales de los usuarios.
