# Dominio de Recursos - Javier Moreno

## Análisis y resumen del video 8

**Asignatura:** Ingeniería de Proyectos

### Propósito
Comprender cómo identificar, organizar, asignar y nivelar los recursos de un proyecto, y cómo las decisiones sobre recursos pueden afectar el cronograma y la ruta crítica.

## 1. Resumen del video

El video presenta el **Dominio de Recursos** y explica que un proyecto necesita diferentes tipos de recursos para poder realizar su trabajo.

Se destacan seis temas principales:

1. **Dominio de Recursos.**
2. **Recursos humanos y virtuales.**
3. **RBS y calendarios.**
4. **Asignación de tareas.**
5. **Nivelación de recursos.**
6. **Impacto en la ruta crítica.**

El video muestra que administrar recursos no consiste solamente en asignar personas a tareas. También es necesario considerar su disponibilidad, los calendarios, los recursos físicos o tecnológicos, las restricciones y el efecto que una asignación puede tener sobre el cronograma.

## 2. ¿Qué entendemos por recursos?

Los recursos son los elementos necesarios para realizar el trabajo del proyecto.

Pueden incluir:

- Personas.
- Equipos.
- Materiales.
- Instalaciones.
- Tecnología.
- Herramientas.
- Recursos virtuales.

En un proyecto de software, por ejemplo, pueden participar:

- Analistas.
- Diseñadores.
- Desarrolladores.
- Probadores.
- Gerente del proyecto.
- Servidores.
- Computadores.
- Servicios en la nube.

La pregunta básica es:

**¿Qué necesitamos para realizar cada actividad?**

## 3. Recursos humanos y virtuales

El video diferencia entre **recursos humanos y recursos virtuales**.

### Recursos humanos

Son las personas que realizan o apoyan el trabajo.

Ejemplos:

- Desarrollador.
- Diseñador.
- Analista.
- Tester.
- Líder técnico.

### Recursos virtuales

Son recursos tecnológicos o digitales que permiten realizar el trabajo.

Ejemplos:

- Servidores.
- Servicios en la nube.
- Plataformas.
- Herramientas de desarrollo.
- Bases de datos.
- Sistemas de almacenamiento.

Ambos pueden ser necesarios para completar correctamente una actividad.

## 4. Estructura de Desglose de Recursos (RBS)

El video presenta la **RBS — Resource Breakdown Structure**, o **Estructura de Desglose de Recursos**.

La RBS permite organizar los recursos de manera jerárquica.

Por ejemplo:

**Recursos del proyecto**
- Humanos
  - Desarrollo
  - Diseño
  - Pruebas
- Tecnológicos
  - Servidores
  - Bases de datos
  - Herramientas
- Materiales
  - Equipos
  - Otros recursos físicos

Esta estructura facilita identificar qué recursos existen y cómo se agrupan.

### Relación con la EDT

La **EDT** organiza el trabajo.

La **RBS** organiza los recursos.

Por eso ambas estructuras pueden utilizarse conjuntamente para planificar el proyecto.

## 5. Calendarios de recursos

No todas las personas y recursos están disponibles durante todo el tiempo.

El video destaca la importancia de los **calendarios de recursos**.

Un calendario puede indicar:

- Días laborales.
- Horarios.
- Vacaciones.
- Ausencias.
- Disponibilidad parcial.
- Fechas de mantenimiento.
- Restricciones de equipos o servicios.

### Ejemplo

Si un desarrollador solamente está disponible de lunes a jueves, no sería correcto planificar su trabajo como si estuviera disponible todos los días.

Por eso la disponibilidad real debe incorporarse al cronograma.

## 6. Asignación de recursos a tareas

Después de identificar los recursos, es necesario asignarlos a las actividades.

Ejemplo:

| Actividad | Recurso |
|---|---|
| Análisis | Analista |
| Diseño de interfaz | Diseñador |
| Programación | Desarrollador |
| Pruebas | Tester |
| Despliegue | Administrador / recurso técnico |

La asignación debe considerar:

- Capacidad.
- Disponibilidad.
- Conocimientos.
- Calendario.
- Carga de trabajo.

Una persona no debería ser asignada a más trabajo del que realmente puede realizar en el mismo periodo.

## 7. Sobrecarga de recursos

Puede ocurrir que un mismo recurso sea asignado a varias actividades simultáneamente.

Por ejemplo:

**Desarrollador Juan**

- Actividad A: 8 horas.
- Actividad B: 8 horas.
- Actividad C: 8 horas.

Si todas deben realizarse el mismo día, existiría una carga superior a la disponibilidad normal de una sola persona.

Esto genera una **sobrecarga de recursos**.

El gerente del proyecto debe detectar estas situaciones antes de asumir que el cronograma es viable.

## 8. Nivelación de recursos

La **nivelación de recursos** busca ajustar el cronograma cuando la demanda de recursos supera su disponibilidad.

Puede implicar:

- Cambiar fechas.
- Mover actividades.
- Distribuir el trabajo.
- Ajustar asignaciones.
- Utilizar recursos alternativos.
- Modificar la secuencia cuando sea posible.

El objetivo es lograr un uso más equilibrado de los recursos disponibles.

### Importante

La nivelación puede cambiar la duración del proyecto.

Por eso no debe hacerse de manera aislada: hay que revisar nuevamente el cronograma después de realizar los ajustes.

## 9. Efectos de la nivelación

El video destaca tres efectos importantes de la nivelación:

- Equilibra la demanda y la oferta disponible.
- Puede ajustar fechas de inicio y finalización.
- Es útil cuando existen recursos compartidos o críticos.

Sin embargo, una nivelación mal planteada puede **consumir la holgura** de determinadas actividades.

Cuando se consume la holgura, una actividad que anteriormente tenía margen puede acercarse a la ruta crítica.

Por eso siempre debe revisarse el impacto de los cambios.

## 10. Recursos y ruta crítica

Existe una relación importante entre los recursos y la ruta crítica.

Una modificación en la asignación de recursos puede producir cambios en el cronograma.

Por ejemplo:

**Menor disponibilidad de un recurso**
↓
**Actividad se retrasa**
↓
**Se consume holgura**
↓
**Puede cambiar la ruta crítica**
↓
**Puede cambiar la fecha final**

Por esta razón, gestionar recursos también significa proteger la viabilidad del cronograma.

## 11. La intuición del gerente de proyectos

El cierre del video presenta una idea importante: además de los cálculos y herramientas, existe un componente de **criterio y experiencia humana**.

La planificación puede indicar que un recurso está disponible, pero el gerente debe considerar también aspectos reales como:

- Carga de trabajo.
- Complejidad.
- Fatiga.
- Coordinación.
- Experiencia.
- Bienestar del equipo.
- Riesgos.

La idea presentada es que la intuición humana puede ayudar a equilibrar las exigencias del proyecto con el bienestar y las condiciones reales del equipo.

Esto no significa reemplazar los datos, sino complementar la información con conocimiento del contexto.

## 12. Análisis del video

El principal aprendizaje es que **tener recursos disponibles no significa automáticamente que estén correctamente utilizados**.

Es necesario conocer:

**Qué recursos tenemos → Cuándo están disponibles → Qué trabajo deben realizar → Cuánta capacidad tienen → Qué impacto produce su asignación**

La gestión de recursos está directamente relacionada con el cronograma.

Un proyecto puede tener todas sus actividades correctamente definidas, pero si las personas o recursos necesarios no están disponibles en el momento adecuado, el cronograma puede dejar de ser realista.

## 15. Ejemplo práctico

Supongamos un proyecto con un único desarrollador.

| Actividad | Duración | Recurso |
|---|---:|---|
| A - Login | 2 días | Desarrollador |
| B - Usuarios | 3 días | Desarrollador |
| C - Reportes | 2 días | Desarrollador |

Si A, B y C requieren al mismo desarrollador al mismo tiempo, existe una limitación de capacidad.

Una alternativa sería organizar las actividades:

**A → B → C**

o incorporar otro recurso si las condiciones del proyecto lo permiten.

La decisión debe considerar tiempo, costo, disponibilidad y efecto sobre el cronograma.

## 16. Importancia para Ingeniería de Proyectos

El dominio de recursos es fundamental porque permite convertir un cronograma teórico en un plan más cercano a la realidad.

Para un proyecto de software es necesario conocer:

- Quién realiza cada actividad.
- Qué conocimientos necesita.
- Cuándo está disponible.
- Qué herramientas necesita.
- Qué otros trabajos tiene asignados.
- Qué recursos tecnológicos requiere.

La correcta gestión de recursos ayuda a prevenir retrasos y permite tomar decisiones antes de que los problemas afecten la fecha final.

## 17. Conclusión

El video muestra que los recursos son una parte esencial de la ejecución del proyecto.

La gestión puede resumirse así:

**Identificar → Organizar → Asignar → Revisar disponibilidad → Nivelar → Evaluar impacto**

La RBS ayuda a organizar los recursos, los calendarios permiten conocer su disponibilidad y la nivelación ayuda a resolver conflictos de asignación.

La enseñanza más importante es que los recursos están conectados con el cronograma: una decisión sobre recursos puede consumir holgura, afectar la ruta crítica y modificar la fecha final del proyecto.

Por eso la gestión de recursos debe combinar información, planificación y criterio sobre las condiciones reales del equipo.

## 18. Actividad de análisis

Para un proyecto de software identifica:

### Recursos humanos
- 3 roles necesarios.

### Recursos tecnológicos
- 3 recursos necesarios.

### Asignación
Relaciona cada recurso con una actividad.

### Disponibilidad
Indica cuándo estaría disponible cada recurso.

### Nivelación
Imagina que uno de los recursos tiene dos actividades simultáneas.

Responde:

1. ¿Existe sobrecarga?
2. ¿Qué actividad moverías?
3. ¿Qué efecto podría tener sobre el cronograma?
4. ¿Podría afectar la ruta crítica?
5. ¿Sería necesario utilizar otro recurso?
