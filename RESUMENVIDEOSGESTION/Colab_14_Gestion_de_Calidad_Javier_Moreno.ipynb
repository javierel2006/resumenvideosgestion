### Propósito
Comprender qué significa gestionar la calidad de un proyecto y cómo prevenir errores para entregar resultados que cumplan los requisitos establecidos.

## 1. Resumen del tema

La calidad consiste en que el producto o resultado cumpla los requisitos y necesidades definidos para el proyecto.

En un proyecto de software, la calidad no significa solamente que la aplicación se vea bien. También debe funcionar correctamente, ser segura, mantener un buen rendimiento y cumplir lo solicitado.

La calidad debe trabajarse durante todo el proyecto y no únicamente al final.

## 2. Calidad y requisitos

Los requisitos sirven como referencia para determinar qué debe cumplir el producto.

Por ejemplo, una aplicación puede tener estos requisitos:

- Permitir iniciar sesión.
- Registrar usuarios.
- Responder rápidamente.
- Proteger la información.
- Mostrar correctamente los datos.

Si la aplicación cumple estos requisitos, existe una base para evaluar su calidad.

## 3. Prevenir es mejor que corregir

Una idea importante de la gestión de calidad es que resulta preferible **prevenir los errores** antes que encontrarlos cuando el producto ya está terminado.

Ejemplo:

Si un requisito está mal entendido, descubrirlo durante la planificación puede requerir una corrección pequeña.

Si el mismo error se descubre después de programar y probar todo el módulo, la corrección puede requerir mucho más tiempo y dinero.

## 4. Aseguramiento y control de calidad

### Aseguramiento de la calidad

Se enfoca en los procesos utilizados para desarrollar el proyecto.

Ejemplos:

- Revisiones.
- Estándares.
- Procedimientos.
- Buenas prácticas.
- Revisiones de código.

### Control de calidad

Se enfoca en comprobar los resultados obtenidos.

Ejemplos:

- Pruebas funcionales.
- Pruebas de rendimiento.
- Inspecciones.
- Validación de resultados.

Ambos enfoques se complementan.

## 5. Calidad en un proyecto de software

Algunos aspectos que pueden evaluarse son:

**Funcionalidad:** la aplicación hace lo que debe hacer.

**Usabilidad:** los usuarios pueden utilizarla fácilmente.

**Rendimiento:** responde de forma adecuada.

**Seguridad:** protege la información y los accesos.

**Mantenibilidad:** puede modificarse y mantenerse.

**Compatibilidad:** funciona en los entornos definidos.

La importancia de cada característica depende de los requisitos del proyecto.

## 6. Costos de calidad

La calidad también tiene relación con los costos.

Se pueden considerar:

- Costos de prevención.
- Costos de evaluación o comprobación.
- Costos por fallas internas.
- Costos por fallas externas.

Por ejemplo, capacitar al equipo puede representar un costo de prevención, mientras que corregir un error después de entregar el software puede convertirse en un costo por falla externa.

## 7. Ejemplo sencillo

Supongamos que un proyecto tiene un presupuesto de **$10.000.000**.

Durante las pruebas se encuentran errores antes de entregar el producto.

Si el equipo corrige los errores internamente, se evita que lleguen al cliente.

Pero si esos errores llegan a producción, pueden generar:

- Retrabajo.
- Reclamos.
- Pérdida de tiempo.
- Costos adicionales.
- Afectación de la confianza del usuario.

Por eso las pruebas y revisiones durante el desarrollo son importantes.

## 8. Indicadores de calidad

Los indicadores permiten medir el comportamiento de la calidad.

Algunos ejemplos son:

- Número de errores encontrados.
- Errores por módulo.
- Porcentaje de pruebas exitosas.
- Tiempo promedio de solución.
- Porcentaje de requisitos cumplidos.
- Incidencias después de la entrega.

Los indicadores deben utilizarse para entender el estado del proyecto y tomar acciones.

```python
# Ejemplo sencillo de porcentaje de pruebas exitosas

pruebas_totales = 100
pruebas_exitosas = 92

porcentaje = (pruebas_exitosas / pruebas_totales) * 100

print(f"Pruebas exitosas: {porcentaje:.1f}%")
```

## 9. Gestión de defectos

Cuando se encuentra un defecto, es útil registrar:

1. Qué ocurrió.
2. Dónde ocurrió.
3. Cómo reproducirlo.
4. Qué impacto tiene.
5. Quién lo atenderá.
6. Estado de la solución.
7. Resultado de la nueva prueba.

Un defecto no debe simplemente marcarse como corregido. Es recomendable comprobar que la solución realmente funciona.

## 10. Calidad y mejora continua

La información obtenida durante el proyecto puede utilizarse para mejorar el proceso.

Una secuencia sencilla es:

**Planificar → Ejecutar → Revisar → Mejorar**

Por ejemplo, si las pruebas encuentran repetidamente el mismo tipo de error, el equipo puede revisar el proceso de desarrollo para evitar que vuelva a ocurrir.

## 12. Ejemplo aplicado a una aplicación web

Supongamos que una aplicación tiene un módulo de inicio de sesión.

Los requisitos indican que:

- El usuario debe ingresar correo y contraseña.
- Las credenciales incorrectas deben ser rechazadas.
- La contraseña no debe mostrarse.

- Datos correctos.
- Datos incorrectos.

- Contraseñas incorrectas.

Esto permite verificar que el resultado cumple lo esperado.

## 14. Importancia para Ingeniería de Proyectos

La calidad no debe considerarse una actividad que se realiza únicamente antes de entregar un proyecto.

Debe estar presente desde los requisitos hasta el cierre.

Una gestión adecuada ayuda a reducir retrabajos, detectar problemas temprano y aumentar la posibilidad de entregar un producto que cumpla lo acordado.

## 15. Conclusión

La gestión de calidad busca que el proyecto produzca resultados que cumplan los requisitos establecidos.

Para lograrlo es necesario:

**Definir requisitos → Planificar la calidad → Ejecutar correctamente → Revisar → Corregir → Mejorar**

En software, las revisiones y pruebas durante el desarrollo permiten detectar problemas antes de que lleguen al usuario.
