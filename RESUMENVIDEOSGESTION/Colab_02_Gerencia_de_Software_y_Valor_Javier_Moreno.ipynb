# Gerencia de Software y Valor - Javier Moreno

## Análisis y resumen del video 2

**Asignatura:** Ingeniería de Proyectos

### Propósito
Comprender cómo la gerencia de proyectos de software se relaciona con la generación de valor, la estrategia, el caso de negocio y el acta constitutiva.

## 1. Resumen del video

El video explica la relación entre la **gerencia de software y la generación de valor**.

La presentación parte de la evolución de la gestión de proyectos y muestra que el enfoque actual busca conectar el proyecto con las necesidades de la organización y con los resultados que se esperan obtener.

El contenido se organiza alrededor de cinco temas:

1. **Estructura del PMBOK 8.ª edición.**
2. **Portafolios y operación.**
3. **Sistema de valor.**
4. **Tutorial de viabilidad.**
5. **Caso de negocio y acta constitutiva.**

Una idea central es que un proyecto no debe verse solamente como un conjunto de actividades técnicas. Debe existir una razón para realizarlo, unos objetivos medibles y una relación clara entre la inversión realizada y el valor que se espera obtener.

## 2. Estructura del PMBOK 8.ª edición

El video presenta el PMBOK como una guía para orientar la gestión de proyectos y no simplemente como una lista rígida de pasos.

La gestión debe conectar:

**Estrategia → Proyecto → Entregables → Resultados → Valor**

Esto permite comprender que las actividades del proyecto tienen sentido cuando contribuyen a un objetivo mayor.

### Idea sencilla

Un proyecto de software puede entregar una aplicación funcional, pero eso por sí solo no garantiza que haya generado valor.

Por ejemplo, una aplicación puede funcionar técnicamente y, aun así, no solucionar correctamente la necesidad que originó el proyecto.

Por eso es necesario considerar desde el principio:

- Qué necesidad existe.
- Qué objetivo se quiere alcanzar.
- Qué resultado se espera.
- Cómo se comprobará el éxito.
- Qué beneficios se esperan obtener.

## 3. Portafolios, operación y sistema de valor

### Portafolio

Un portafolio permite observar diferentes proyectos y trabajos de una organización desde una perspectiva más amplia.

La pregunta no es únicamente:

**¿El proyecto está funcionando?**

También interesa saber:

**¿Este proyecto contribuye a la estrategia y a los objetivos de la organización?**

### Operación

La operación corresponde al funcionamiento habitual de la organización.

Un proyecto es temporal, mientras que las operaciones normalmente continúan.

En software, por ejemplo:

- Crear un nuevo sistema puede ser un **proyecto**.
- Mantener y utilizar ese sistema diariamente forma parte de la **operación**.

### Sistema de valor

El video presenta la conexión entre estrategia, trabajo de la organización, proyectos y resultados.

La idea fundamental es que la gestión de proyectos debe contribuir a producir resultados útiles y beneficios, no limitarse a completar tareas.

## 4. Viabilidad del proyecto

El video introduce el **análisis de viabilidad**.

Antes de invertir recursos importantes en un proyecto es necesario estudiar si la propuesta tiene sentido.

Entre los aspectos que pueden analizarse están:

- La necesidad que se quiere solucionar.
- Los objetivos.
- Los beneficios esperados.
- Los recursos necesarios.
- Los costos.
- La posibilidad de realizar el proyecto.
- La relación entre inversión y beneficio.

### Ejemplo

Supongamos que una empresa quiere desarrollar un sistema para automatizar una actividad que actualmente se realiza manualmente.

Antes de comenzar a programar sería necesario analizar:

**Necesidad → Alternativas → Costos → Beneficios → Viabilidad → Decisión**

Esto evita iniciar proyectos sin conocer suficientemente su propósito y sus posibles resultados.

## 5. Caso de negocio

El **caso de negocio** permite justificar una propuesta de proyecto desde una perspectiva organizacional y económica.

En el video se presenta como un estudio de viabilidad económica que ayuda a establecer la validez de los beneficios de un proyecto.

Puede considerar elementos como:

- Necesidades comerciales.
- Alineación estratégica.
- Beneficios financieros.
- Análisis de viabilidad.
- Costos e inversión.
- Resultados esperados.

### Pregunta fundamental

**¿Tiene sentido realizar este proyecto teniendo en cuenta lo que cuesta y lo que se espera obtener?**

El caso de negocio ayuda a responder esta pregunta antes de comprometer completamente los recursos.

## 6. Acta constitutiva del proyecto

Otro elemento importante presentado es el **Project Charter o Acta Constitutiva**.

Es el documento que autoriza formalmente el proyecto y proporciona una base para iniciar su gestión.

De forma sencilla:

**Caso de negocio → decisión de realizar el proyecto → Acta Constitutiva → inicio formal del proyecto**

El acta ayuda a establecer elementos como:

- Propósito del proyecto.
- Objetivos.
- Límites iniciales del alcance.
- Recursos iniciales.
- Interesados relevantes.
- Criterios de éxito o salida.
- Autoridad para comenzar el proyecto.

### Diferencia importante

El **caso de negocio** ayuda a justificar por qué conviene realizar el proyecto.

El **acta constitutiva** formaliza la autorización para iniciar el proyecto.

## 8. Análisis del video

El principal aporte del video es mostrar que la gerencia de un proyecto de software tiene una dimensión que va más allá de la parte técnica.

Un proyecto puede tener buenos programadores, utilizar herramientas modernas y entregar una aplicación funcional, pero todavía es necesario comprobar que el resultado responde a una necesidad real.

Por eso la gestión debe conectar tres aspectos:

### 1. Necesidad
¿Qué problema o situación requiere una solución?

### 2. Gestión
¿Cómo se organizará el trabajo para desarrollar la solución?

### 3. Valor
¿Qué beneficio o resultado útil se espera obtener?

Desde esta perspectiva, la tecnología es un medio para alcanzar resultados y no el único objetivo del proyecto.

## 9. Relación con Ingeniería de Proyectos

Este contenido es especialmente importante para Ingeniería de Proyectos porque permite comprender que antes de planificar actividades detalladas hay que tener claro **qué se quiere lograr y por qué**.

En un proyecto de software:

- El problema define la necesidad.
- Los objetivos orientan el trabajo.
- El caso de negocio ayuda a justificar la inversión.
- La viabilidad permite analizar si la propuesta puede realizarse.
- El acta constitutiva autoriza formalmente el proyecto.
- Los entregables representan productos concretos del trabajo.
- Los resultados y beneficios permiten evaluar el valor generado.

## 10. Conceptos para recordar

| Concepto | Explicación sencilla |
|---|---|
| Proyecto | Trabajo temporal realizado para alcanzar un resultado |
| Portafolio | Conjunto de proyectos y trabajos gestionados considerando objetivos organizacionales |
| Operación | Actividades habituales y continuas de la organización |
| Sistema de valor | Conexión entre estrategia, trabajo, resultados y beneficios |
| Viabilidad | Análisis para determinar si una propuesta puede y tiene sentido realizarse |
| Caso de negocio | Justificación de la inversión y de los beneficios esperados |
| Acta constitutiva | Documento que autoriza formalmente el proyecto |
| Valor | Beneficio o resultado útil producido por el trabajo |

## 11. Conclusión

El video permite comprender que la gerencia de software debe comenzar con una pregunta fundamental:

**¿Qué valor se espera generar con el proyecto?**

A partir de esa pregunta se pueden definir la necesidad, los objetivos, el análisis de viabilidad, el caso de negocio y el acta constitutiva.

La principal enseñanza es que un proyecto de software no debe evaluarse únicamente por haber terminado el desarrollo. También debe analizarse si produjo el resultado y los beneficios que justificaron su realización.

Este tema prepara el camino para los siguientes videos, donde se profundizará en la gobernanza, los requisitos, el alcance y la organización concreta del trabajo.

## 12. Actividad de análisis

Piensa en un proyecto sencillo de desarrollo de software, por ejemplo, una aplicación para una empresa.

Completa mentalmente esta secuencia:

**Necesidad → Objetivo → Beneficio → Viabilidad → Caso de negocio → Acta constitutiva**

La finalidad es comprobar que antes de comenzar el desarrollo existe una relación lógica entre el problema que se quiere resolver y el valor que se espera obtener.
