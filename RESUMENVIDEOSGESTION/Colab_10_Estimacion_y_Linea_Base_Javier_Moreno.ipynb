# Estimación y Línea Base - Javier Moreno

## Análisis y resumen del video 10

**Asignatura:** Ingeniería de Proyectos

### Propósito
Comprender cómo se construye una estimación de costos, cómo se incorporan las reservas y cómo se establece una **línea base de costos** para controlar financieramente el proyecto.

## 1. Resumen del video

El video explica cómo pasar de las estimaciones de costos a una **línea base aprobada** que permita controlar el desempeño financiero del proyecto.

Los temas principales son:

- Estimación de costos.
- Reservas para contingencias.
- Reservas de gestión.
- Diagrama o curva S.
- Costo de la calidad.
- Consolidación de estimaciones.
- Línea base de costos.
- Control de cambios.

La idea central es:

**Estimaciones → Reservas → Presupuesto aprobado → Línea base → Seguimiento y control**

La línea base funciona como una referencia contra la cual se puede comparar el comportamiento real del proyecto.

## 2. Estimación de costos

La estimación de costos busca determinar cuánto dinero puede requerir el trabajo del proyecto.

Para realizarla pueden considerarse:

- Actividades.
- Recursos.
- Duraciones.
- Materiales.
- Tecnología.
- Personal.
- Información histórica.
- Experiencia del equipo.

En un proyecto de software, por ejemplo, puede estimarse por separado el costo de:

- Diseño.
- Backend.
- Desarrollo de modelos.
- Pruebas.
- Infraestructura.

Después se consolidan las estimaciones para obtener una visión económica del proyecto.

## 3. Reservas del proyecto

El video diferencia dos tipos de reservas importantes.

### Reserva de contingencia

Se utiliza para responder a **riesgos conocidos** que ya han sido identificados y para los cuales existe una respuesta prevista.

Ejemplo:

Si se conoce que una determinada actividad puede requerir trabajo adicional por un riesgo identificado, puede contemplarse una reserva de contingencia.

### Reserva de gestión

Está destinada a **imprevistos o incertidumbres desconocidas**, bajo la autoridad de la dirección o gestión correspondiente.

### Diferencia sencilla

**Contingencia → riesgos conocidos**

**Gestión → imprevistos desconocidos**

Esta diferencia es fundamental para comprender cómo se construye el presupuesto total del proyecto.

## 4. Ejemplo de consolidación mostrado en el video

El video presenta un ejemplo de estimación por componentes:

| Componente | Costo | Reserva | Total |
|---|---:|---:|---:|
| Diseño | $10K | $2K | $12K |
| Backend | $20K | $3K | $23K |
| Modelo IA | $30K | $5K | $35K |
| **Total** | **$60K** | **$10K** | **$70K** |

El ejemplo muestra cómo las estimaciones de los componentes pueden consolidarse junto con sus reservas para obtener una estructura financiera del proyecto.

En este ejemplo, la suma de las estimaciones es **$60K** y las reservas mostradas suman **$10K**, dando un total de **$70K**.

Estos valores corresponden al ejemplo presentado en el material del video y no deben interpretarse como costos reales de un proyecto.

## 5. ¿Qué es la línea base de costos?

La **línea base de costos** es la versión aprobada de los costos planificados que se utilizará como referencia para controlar el proyecto.

El video destaca que:

- Contiene estimaciones aprobadas.
- Incluye las reservas de contingencia.
- Excluye las reservas de gestión.
- Requiere control de cambios.

### Idea sencilla

La línea base responde:

**¿Cuál es el costo aprobado que esperamos utilizar según el plan?**

Después, durante la ejecución, los costos reales pueden compararse con esta referencia.

## 6. Presupuesto y línea base

Es importante no confundir presupuesto con línea base.

De forma sencilla:

**Estimaciones**
↓
**Reservas de contingencia**
↓
**Línea base de costos**

La reserva de gestión se mantiene fuera de la línea base de costos, aunque forma parte de la visión más amplia de los fondos autorizados para el proyecto.

Esto permite distinguir entre el costo planificado que se utiliza para medir el desempeño y los fondos de gestión que pueden utilizarse bajo el control correspondiente.

## 7. Curva S

El video presenta el **Diagrama o Curva S**.

Es un gráfico que muestra los **costos acumulados durante un periodo de tiempo**.

Se denomina Curva S porque, en muchos proyectos, el crecimiento acumulado del costo adopta aproximadamente esa forma.

Puede utilizarse para observar:

- Cómo se distribuyen los costos en el tiempo.
- Cuánto se ha planificado gastar hasta una fecha.
- Cómo evoluciona el costo acumulado.

### Ejemplo conceptual

**Inicio → crecimiento lento → mayor ritmo de ejecución → estabilización → cierre**

La Curva S ayuda a visualizar el comportamiento financiero del proyecto a través del tiempo.

## 8. Costo de la calidad

El video también presenta el **costo de la calidad**.

La idea es buscar un equilibrio entre invertir en prevención y evaluación y asumir los costos que pueden aparecer por fallas.

### Prevención

Acciones para evitar errores antes de que ocurran.

Ejemplos:

- Capacitación.
- Buen diseño.
- Revisiones.
- Pruebas planificadas.

### Evaluación

Acciones para comprobar si el producto cumple los requisitos.

Ejemplos:

- Pruebas.
- Inspecciones.
- Revisiones.

### Fallas

Costos producidos por errores o problemas.

Ejemplos:

- Retrabajo.
- Corrección de errores.
- Incidentes.
- Problemas después de la entrega.

La idea central es que invertir adecuadamente en prevención y evaluación puede ayudar a evitar costos posteriores asociados con fallas.

## 9. Relación entre costo y calidad

En un proyecto de software puede existir una situación como esta:

**Poca prevención**
→ más errores
→ más retrabajo
→ mayores costos de falla

Mientras que una inversión adecuada en:

**Prevención + Evaluación**
→ puede reducir errores
→ puede disminuir retrabajo
→ puede proteger el presupuesto

Por eso el costo de la calidad no debe entenderse simplemente como un gasto adicional. Debe analizarse en relación con los costos que pueden aparecer cuando la calidad no se controla adecuadamente.

## 10. Análisis del video

La principal enseñanza del video es que estimar costos no consiste solamente en sumar precios.

Es necesario construir una estructura que contemple:

**Trabajo → Costos → Riesgos → Reservas → Presupuesto → Línea base**

La línea base permite tener una referencia aprobada para el control.

Además, la Curva S ayuda a observar la evolución acumulada de los costos.

Por otro lado, el costo de la calidad recuerda que las decisiones de prevención y evaluación pueden tener consecuencias económicas sobre el proyecto.

En conjunto, estos conceptos permiten pasar de una estimación inicial a un sistema más organizado para controlar los costos.

## 13. Ejemplo sencillo de línea base

Supongamos que un proyecto tiene las siguientes estimaciones:

- Diseño: $10.000.000
- Backend: $20.000.000
- Modelo de IA: $30.000.000

Total de estimaciones:

**$60.000.000**

Si las reservas de contingencia consideradas son:

- Diseño: $2.000.000
- Backend: $3.000.000
- Modelo IA: $5.000.000

Total de contingencias:

**$10.000.000**

Entonces:

**Línea base de costos = $60.000.000 + $10.000.000**

**Línea base = $70.000.000**

Este ejercicio reproduce el ejemplo mostrado en el video.

## 14. Importancia para Ingeniería de Proyectos

La línea base es importante porque proporciona una referencia para el seguimiento financiero.

Durante la ejecución pueden compararse:

**Costo planificado**

frente a

**Costo real**

Si existen diferencias importantes, el equipo puede investigar las causas y determinar si se requiere alguna acción o cambio aprobado.

Esto conecta posteriormente con el **Valor Ganado**, porque para medir el desempeño económico y temporal se necesita una referencia planificada contra la cual realizar comparaciones.

## 15. Conclusión

El video muestra cómo convertir las estimaciones de costos en una referencia formal para controlar el proyecto.

La secuencia principal es:

**Estimar → Incorporar contingencias → Aprobar → Establecer línea base → Controlar**

También se presentan dos herramientas conceptuales importantes:

**Curva S → permite observar costos acumulados en el tiempo.**

**Costo de la calidad → permite analizar la relación entre prevención, evaluación y fallas.**

La enseñanza principal es que la gestión financiera debe prepararse antes de comenzar el seguimiento, porque necesitamos una referencia aprobada para comparar lo planeado con lo que realmente ocurre.
