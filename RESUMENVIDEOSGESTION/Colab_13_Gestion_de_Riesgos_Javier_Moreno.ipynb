## 1. Resumen del tema

Un riesgo es una situación o evento que puede ocurrir y que, si ocurre, puede afectar los objetivos del proyecto.

El riesgo puede afectar:

- Alcance.
- Tiempo.
- Costos.
- Calidad.
- Recursos.
- Resultados esperados.

La gestión de riesgos busca anticiparse a los problemas y preparar respuestas antes de que ocurran.

## 2. Riesgo y problema no son lo mismo

### Riesgo
Es algo que **podría ocurrir** en el futuro.

Ejemplo:

> Un proveedor podría entregar tarde un servicio necesario para el proyecto.

### Problema
Es algo que **ya está ocurriendo**.

Ejemplo:

> El proveedor ya informó que la entrega tendrá un retraso.

Por eso, los riesgos se gestionan antes de que sucedan y los problemas requieren acciones inmediatas.

## 3. Identificación de riesgos

Para identificar riesgos se pueden utilizar diferentes fuentes:

- Experiencia de proyectos anteriores.
- Reuniones con el equipo.
- Opiniones de especialistas.
- Revisión de requisitos.
- Análisis del cronograma.
- Análisis de costos.
- Dependencias externas.
- Lecciones aprendidas.

La identificación debe realizarse durante todo el proyecto, no solamente al comienzo.

## 4. Registro de riesgos

Una forma sencilla de organizar los riesgos es mediante un registro.

| Riesgo | Probabilidad | Impacto | Respuesta |
|---|---|---|---|
| Retraso del proveedor | Alta | Alto | Tener proveedor alternativo |
| Error técnico | Media | Alto | Realizar pruebas tempranas |
| Falta de personal | Media | Medio | Planificar reemplazos |
| Cambio de requisito | Alta | Medio | Controlar cambios |

Este registro ayuda al equipo a saber qué debe vigilar y qué acciones puede realizar.

## 5. Probabilidad e impacto

Dos elementos importantes son:

**Probabilidad:** posibilidad de que el riesgo ocurra.

**Impacto:** efecto que tendría sobre el proyecto si ocurre.

Un riesgo con probabilidad alta e impacto alto requiere especial atención.

Una forma sencilla de priorizar es utilizar una matriz:

**Nivel de riesgo = Probabilidad × Impacto**

## 6. Ejemplo de cálculo

Supongamos:

- Probabilidad = 0,60
- Impacto económico = $20.000

La exposición económica esperada puede calcularse de forma sencilla:

**Exposición = Probabilidad × Impacto**

**Exposición = 0,60 × $20.000**

**Exposición = $12.000**

Este valor puede ayudar a comparar riesgos y apoyar decisiones sobre las respuestas.

## 7. Respuestas a riesgos negativos

Cuando un riesgo puede afectar negativamente al proyecto, algunas respuestas habituales son:

### Evitar
Cambiar el plan para eliminar la amenaza.

### Mitigar
Reducir la probabilidad o el impacto.

### Transferir
Pasar parte del impacto o responsabilidad a un tercero.

### Aceptar
Reconocer el riesgo y actuar solamente si llega a ocurrir, o establecer una reserva/acción de contingencia cuando corresponda.

## 8. Oportunidades

No todos los riesgos representan amenazas.

También pueden existir oportunidades que produzcan un resultado favorable.

Algunas respuestas pueden ser:

- Aprovechar.
- Potenciar.
- Compartir.
- Aceptar.

Ejemplo:

> El equipo descubre una herramienta que podría reducir el tiempo de desarrollo.

En lugar de tratarla como un problema, se puede analizar como una oportunidad para mejorar el proyecto.

## 9. Riesgos en proyectos de software

En un proyecto de software pueden aparecer riesgos como:

- Requisitos incompletos.
- Cambios frecuentes del cliente.
- Fallas de integración.
- Problemas de seguridad.
- Dependencia de servicios externos.
- Falta de personal especializado.
- Errores de estimación.
- Retrasos en pruebas.
- Problemas de infraestructura.
- Cambios tecnológicos.

Identificar estos riesgos temprano permite preparar alternativas.

```python
# Matriz sencilla de probabilidad e impacto

riesgos = {
    "Retraso del proveedor": (4, 4),
    "Error técnico": (3, 5),
    "Falta de personal": (3, 3),
    "Cambio de requisito": (4, 3)
}

for riesgo, (probabilidad, impacto) in riesgos.items():
    nivel = probabilidad * impacto
    print(f"{riesgo}: Probabilidad={probabilidad}, Impacto={impacto}, Nivel={nivel}")
```

## 10. Plan de respuesta

Identificar un riesgo no es suficiente.

Para cada riesgo importante conviene definir:

1. Qué puede ocurrir.
2. Qué lo puede causar.
3. Qué impacto tendría.
4. Quién será responsable.
5. Qué acción se realizará.
6. Cuándo se revisará.
7. Qué hacer si el riesgo se convierte en realidad.

Así se pasa de simplemente conocer el riesgo a gestionarlo.

## 11. Seguimiento de los riesgos

Los riesgos deben revisarse durante la ejecución del proyecto.

Un riesgo puede:

- Aumentar su probabilidad.
- Disminuir su probabilidad.
- Aumentar su impacto.
- Disminuir su impacto.
- Desaparecer.
- Convertirse en problema.
- Generar un nuevo riesgo.

Por eso el registro de riesgos debe mantenerse actualizado.

## 15. Importancia para Ingeniería de Proyectos

Una buena gestión de riesgos no significa intentar eliminar toda incertidumbre.

Significa conocer las situaciones que pueden afectar el proyecto, priorizarlas y preparar respuestas razonables.

La secuencia puede resumirse así:

**Identificar → Analizar → Priorizar → Responder → Vigilar → Actualizar**
