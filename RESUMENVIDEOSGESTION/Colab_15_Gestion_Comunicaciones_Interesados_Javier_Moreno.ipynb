### Propósito

## 1. Resumen del tema

En un proyecto participan o se ven afectados diferentes grupos y personas.

Estos pueden ser:

- Cliente.
- Usuarios.
- Patrocinador.
- Gerente del proyecto.
- Equipo de trabajo.
- Proveedores.
- Directivos.
- Entidades externas.

Cada interesado puede tener necesidades, expectativas y niveles de participación diferentes.

La gestión de interesados busca conocer esas necesidades y establecer una relación adecuada durante el proyecto.

## 2. ¿Qué es un interesado?

Un interesado es una persona, grupo u organización que puede afectar el proyecto, verse afectado por él o tener algún interés relacionado con sus resultados.

Ejemplo en un proyecto de software:

**Cliente:** solicita el sistema.

**Usuarios:** utilizan el sistema.

**Equipo de desarrollo:** construye la solución.

**Gerente del proyecto:** coordina el trabajo.

**Proveedor:** puede proporcionar infraestructura o servicios.

Todos pueden tener una relación diferente con el proyecto.

## 3. Identificación de interesados

Al iniciar el proyecto es importante identificar:

1. Quiénes participan.
2. Qué necesitan.
3. Qué esperan.
4. Qué influencia tienen.
5. Cómo pueden afectar el proyecto.
6. Qué información necesitan recibir.

Esta información puede organizarse en un registro de interesados.

## 4. Influencia e interés

Una herramienta sencilla es clasificar a los interesados según:

**Nivel de interés:** cuánto les importa el proyecto.

**Nivel de influencia:** cuánto pueden afectar las decisiones o resultados.

De manera general:

| Interés | Influencia | Estrategia |
|---|---|---|
| Alto | Alta | Gestionar de cerca |
| Alto | Baja | Mantener informados |
| Bajo | Alta | Mantener satisfechos |
| Bajo | Baja | Vigilar |

La clasificación puede cambiar durante el proyecto.

## 5. Comunicación del proyecto

La comunicación permite compartir información de manera organizada.

Algunas preguntas importantes son:

- ¿Qué información se debe comunicar?
- ¿Quién necesita recibirla?
- ¿Cuándo debe recibirla?
- ¿Cómo se enviará?
- ¿Quién será responsable?
- ¿Qué nivel de detalle necesita cada persona?

No todos los interesados necesitan recibir la misma información.

## 6. Canales de comunicación

Dependiendo del proyecto se pueden utilizar:

- Reuniones.
- Correo electrónico.
- Informes.
- Videollamadas.
- Chats de trabajo.
- Tableros de tareas.
- Documentos compartidos.
- Presentaciones.

La elección depende del tipo de información, urgencia y características de los interesados.

## 7. Comunicación efectiva

Una comunicación efectiva debe ser:

**Clara:** fácil de entender.

**Oportuna:** entregada cuando se necesita.

**Completa:** contiene la información necesaria.

**Concreta:** evita información innecesaria.

**Adecuada:** utiliza el canal y nivel de detalle correcto.

Un mensaje incorrecto o tardío puede generar confusión y afectar el proyecto.

## 8. Ejemplo sencillo

Supongamos que un proyecto de software presenta un retraso de una semana.

El equipo técnico puede necesitar información detallada sobre las causas y las tareas afectadas.

El cliente puede necesitar conocer:

- Nuevo plazo.
- Funcionalidades afectadas.
- Impacto esperado.
- Plan de recuperación.

La dirección puede necesitar:

- Estado general.
- Impacto económico.
- Riesgos.
- Decisiones requeridas.

La información se adapta al interesado.

## 9. Plan de comunicaciones

Un plan sencillo puede organizarse así:

| Interesado | Información | Frecuencia | Medio | Responsable |
|---|---|---|---|---|
| Cliente | Avance | Semanal | Reunión | Gerente |
| Equipo | Tareas | Diaria | Tablero | Líder |
| Dirección | Estado general | Mensual | Informe | Gerente |
| Usuarios | Cambios | Según necesidad | Correo | Analista |

Este tipo de planificación evita que la comunicación dependa de la improvisación.

```python
# Ejemplo sencillo de registro de interesados

interesados = [
    ("Cliente", "Alto", "Alto"),
    ("Usuarios", "Alto", "Medio"),
    ("Equipo de desarrollo", "Alto", "Alto"),
    ("Proveedor", "Medio", "Medio"),
    ("Dirección", "Medio", "Alto")
]

for nombre, interes, influencia in interesados:
    print(f"{nombre}: interés={interes}, influencia={influencia}")
```

## 10. Gestión de expectativas

Las expectativas deben gestionarse desde el inicio.

Es importante diferenciar entre:

- Lo que el proyecto realmente puede entregar.
- Lo que el cliente espera.
- Lo que está incluido en el alcance.
- Lo que requiere un cambio formal.

Una comunicación clara ayuda a evitar promesas que no están respaldadas por el alcance, tiempo o presupuesto.

## 11. Conflictos y comunicación

Los conflictos pueden aparecer por:

- Diferencias de prioridades.
- Cambios de requisitos.
- Problemas de recursos.
- Retrasos.
- Diferencias de interpretación.
- Falta de información.

Una comunicación organizada permite identificar estos problemas antes de que afecten seriamente el proyecto.

## 13. Ejemplo aplicado a software

En un proyecto de aplicación web, el cliente solicita una nueva función que no estaba contemplada.

El gerente del proyecto debe:

1. Registrar la solicitud.
2. Analizar su impacto.
3. Revisar tiempo y costo.
4. Informar a los interesados correspondientes.
5. Obtener la aprobación necesaria.
6. Actualizar el plan si el cambio es aprobado.
7. Comunicar la decisión al equipo.

La comunicación acompaña todo el proceso de cambio.

## 15. Importancia para Ingeniería de Proyectos

Un proyecto puede tener una buena planificación técnica y aun así presentar dificultades si la comunicación falla.

Por eso es necesario:

**Identificar → Analizar → Planificar → Comunicar → Escuchar → Actualizar**

La comunicación no consiste solamente en enviar información. También implica recibir comentarios, entender necesidades y gestionar expectativas.

## 16. Conclusión

La gestión de interesados permite conocer quiénes están relacionados con el proyecto y cómo deben ser involucrados.

La gestión de comunicaciones permite entregar la información adecuada a la persona adecuada, en el momento adecuado y mediante el canal apropiado.

Ambas actividades ayudan a mantener alineados al equipo, cliente, usuarios y demás participantes del proyecto.
